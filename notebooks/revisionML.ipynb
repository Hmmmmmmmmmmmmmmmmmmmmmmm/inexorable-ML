## Custom Scripts

### ws-wallpaper
**Path:** `~/.local/bin/ws-wallpaper`
**Purpose:** Per-workspace wallpaper manager with daemon listener.

```
Usage:
  ws-wallpaper set default <file>      Set default wallpaper
  ws-wallpaper set <1-10> <file>       Set custom wallpaper for workspace
  ws-wallpaper unset <1-10>            Remove custom (falls back to default)
  ws-wallpaper live <ws> <video>       Start mpvpaper video overlay
  ws-wallpaper live-off <ws>           Stop mpvpaper overlay
  ws-wallpaper daemon                  Start workspace listener
  ws-wallpaper status                  Show current config
  ws-wallpaper reload                  Re-preload all wallpapers
```

**How it works:**
- Stores wallpapers at `~/.config/hypr/wallpapers/{default,1,2,3}.png`
- Config file: `~/.config/hypr/wallpapers/wallpapers.conf`
- Daemon listens to Hyprland socket, on workspace change applies the matching wallpaper via `hyprctl hyprpaper wallpaper`
- Preloads all wallpapers at startup for instant switching
- `live` mode launches `mpvpaper --layer overlay` on top of hyprpaper (video overlay, not replacement)

### term-opacity
**Path:** `~/.local/bin/term-opacity`
**Purpose:** Runtime terminal transparency adjustment.

```
Usage:
  term-opacity show                     List terminals
  term-opacity set <0.0-1.0>           Set opacity for ALL terminals
  term-opacity set <0.0-1.0> <pid>     Set opacity for specific PID
  term-opacity set <0.0-1.0> <search>  Set opacity by title match
  term-opacity reset                    Reset to default (0.75)
  term-opacity reset <pid|search>       Reset one terminal
```

**How it works:**
- Uses `hyprctl dispatch setprop address:<addr> alpha <value>`
- Supports multiple terminal types: foot, kitty, alacritty, wezterm, ghostty
- Default opacity: 0.75 (from Hyprland windowrule)
- No `override` flag on windowrule → dynamic changes persist

### splitratio-cycle
**Path:** `~/.local/bin/splitratio-cycle`
**Purpose:** Cycle between split ratios 0.25 → 0.50 → 0.75.
- Caches index at `~/.cache/hypr-split-ratio`
- Sends notification with current ratio

### brightness-up / brightness-down
**Paths:** `~/.local/bin/brightness-up`, `~/.local/bin/brightness-down`
**Purpose:** Increment/decrement screen brightness (used by media keys)

````python
# OPTION 2: Subplots (all plots in a grid)
import math

# Select only numerical columns
num_cols = df.select_dtypes(include='number').columns

n_cols = 4                           # Number of plots per row
n_rows = math.ceil(len(num_cols) / n_cols)

# Create subplot grid
fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5*n_rows))
axes = axes.flatten()                  # Flatten axes for easy looping

# Plot each column in its own subplot
for ax, col in zip(axes, num_cols):
    sns.histplot(df[col], kde=True, ax=ax)
    ax.set_title(col)

# Remove empty subplots
for ax in axes[len(num_cols):]:
    ax.remove()

plt.tight_layout()                     # Adjust spacing
plt.show()
````

````python
#OPTION 4: FacetGrid (Seaborn-managed layout)

# Select only numerical columns
num_cols = df.select_dtypes(include='number').columns

# Reshape data from wide to long format
df_melted = df[num_cols].melt(
    var_name='Feature',
    value_name='Value'
)

# Create FacetGrid
g = sns.FacetGrid(
    df_melted,
    col='Feature',
    col_wrap=4,
    sharex=False,
    sharey=False
)

# Map histogram to each facet
g.map(sns.histplot, 'Value', kde=True)

plt.show()
````
