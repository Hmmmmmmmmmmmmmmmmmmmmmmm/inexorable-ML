A box plot (or box-and-whisker plot) is a way to visualize the distribution of a dataset — showing:

The median (middle value)

The quartiles (spread of the middle 50% of data)

Whiskers (range of most data points)

Outliers (points that fall far from the rest)

It’s great for comparing several datasets at once!

```python
import matplotlib.pyplot as plt
import numpy as np

# Example data
data = np.random.normal(0, 1, 100)  # mean=0, std=1, 100 samples

plt.boxplot(data)
plt.title("Basic Box Plot")
plt.ylabel("Values")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters for the normal distribution
mu = 0      # Mean
sigma = 1   # Standard deviation

# Generate x values
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)

# Calculate the probability density function (PDF)
y = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma)**2)

# Plot the curve
plt.plot(x, y, color='blue', label=f'Normal Distribution\nμ={mu}, σ={sigma}')
plt.title('Normal Distribution')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
```

# Probability Distributions

1. Bernouli's (PMF):
- Definition: The Bernoulli distribution is the simplest discrete probability distribution. It represents the probability
distribution of a random variable that has exactly two possible outcomes: success (with probability p) and failure
(with probability 1 —p). It is used to model binary outcomes, such as a coin flip or a yes/no question.

2. Binomial's (PMF):

Boolean-valued butcome: success (with probability p) or failure

3. Poisson's (PMF):
- In probability theory and statistics, the Poisson distribution is a discrete probability distribution that
expresses the probability of a given number of events occurring in a fixed interval of time if these events
occur with a known constant mean rate and independently of the time since the last event.

7. Log Normal Distribution
- In probability theory, a log-normal (or lognormal) distribution is a continuous
probability distribution of a random variable whose logarithm is normally
distributed. Thus, if the random variable X is log-normally distributed, then Y =
In(X) has a normal distribution.Equivalently, if Y has a normal distribution, then
the exponential function of Y, X = exp(Y), has a log-normal distribution

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python

from scipy.stats import lognorm

# Parameters of the lognormal distribution
mu = 0       # mean of log(X)
sigma = 1    # std dev of log(X)

# Linearly spaced x-values (must be > 0!)
x = np.linspace(0.01, 4, 100)

# Lognormal PDF
pdf = lognorm.pdf(x, s=sigma, scale=np.exp(mu))
pdf2 = lognorm.pdf(x, s=0.5, scale=np.exp(mu))
pdf3 = lognorm.pdf(x, s=0.25, scale=np.exp(mu))
# Plot
# plt.plot(x, x, color='blue')
plt.plot(x, pdf, color='blue')
plt.plot(x, pdf2, color='red')
plt.plot(x, pdf3, color='green')
plt.title("Lognormal Distribution ")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.show()
```

```python

# Linearly spaced array
x = np.linspace(0.1, 10, 100)   # start at 0.1 to avoid log(0)

# Log of the array
y = np.log(x)

plt.plot(x, x, label="x", color="blue")   # original array
plt.plot(x, y, label="log(x)", color="red")  # log of array

plt.title("Array vs Log of Array")
plt.xlabel("x")
plt.ylabel("value")
plt.legend()
plt.grid(True)
plt.show()
```

# Central Limit Theorem
- The central limit theorem relies on the concept of a sampling distribution. which is the
probability distribution of a statistic for a large number of samples taken from a population.
The central limit theorem says that the sampling distribution of the mean will always be normally
distributed, as long as the sample size is large enough. Regardless of whether the population has a
normal. Poisson. binomial. or any other distribution. the sampling distribution of the mean will be
normal.
