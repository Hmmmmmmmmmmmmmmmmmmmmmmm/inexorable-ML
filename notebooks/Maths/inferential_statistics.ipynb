# Inferential Statistics

- Inferential statistics is the branch of statistics that uses sample data to make conclusions, predictions, or decisions about a population.

It allows us to:

- Estimate population parameters (mean, proportion, variance)
- Test hypotheses
- Quantify uncertainty
- Make data-driven decisions

Because we usually cannot measure an entire population, we rely on samples.

| Term           | Meaning                                 |
| -------------- | --------------------------------------- |
| **Population** | Entire group of interest                |
| **Sample**     | Subset of the population                |
| **Parameter**  | Numerical value describing a population |
| **Statistic**  | Numerical value computed from a sample  |


## 3. Main Tools of Inferential Statistics

### A. Estimation
Estimation is used to estimate unknown population parameters based on sample data.

#### (i) Point Estimation
- Provides a single numerical value as an estimate.
- Example:
  - Sample mean x̄ estimates population mean μ
  - Sample proportion p̂ estimates population proportion p

#### (ii) Interval Estimation (Confidence Intervals)
- Provides a range of values within which the population parameter is likely to lie.
- Includes a confidence level (e.g., 90%, 95%, 99%).

Example (Confidence Interval for Mean):
x̄ ± z(α/2) · (σ / √n)

---

### B. Hypothesis Testing
Hypothesis testing is used to test a claim or assumption about a population parameter.

Steps in Hypothesis Testing:
1. State the hypotheses
   - Null hypothesis H₀
   - Alternative hypothesis H₁
2. Choose significance level α (e.g., 0.05)
3. Compute the test statistic
4. Determine the p-value or critical value
5. Make a decision (reject or fail to reject H₀)

---

### C. Prediction
- Uses sample data to predict future or unknown outcomes.
- Commonly applied in regression and forecasting.

---

## 4. Role of the Central Limit Theorem
The Central Limit Theorem (CLT) plays a key role in inferential statistics:

- Sample means follow a normal distribution for large samples
- Enables the use of normal probability models
- Justifies confidence intervals and hypothesis tests
- Applies even when the population distribution is non-normal

---

## 5. Common Inferential Statistical Methods
- Z-test
- t-test
- Chi-square test
- Analysis of Variance (ANOVA)
- Regression Analysis

---

## 6. Assumptions in Inferential Statistics
Inferential methods generally assume:
- Random sampling
- Independence of observations
- Adequate sample size
- Known or estimable population variance

---

## 7. Example
A random sample of 50 students has an average score of 70.

Inferential statistics can be used to:
- Estimate the population mean score
- Test whether the population mean is greater than 65

---

## 8. Descriptive vs Inferential Statistics

Descriptive Statistics:
- Summarizes data
- Uses charts and averages
- Does not generalize beyond data

Inferential Statistics:
- Makes generalizations about the population
- Uses probability theory
- Draws conclusions from sample data


## P Value
- The p value is a number, calculated from a statistical test, that describes how likely you are to have
found a particular set of observations if the null hypothesis were true. P values are used in
hypothesis testing to help decide whether to reject the null hypothesis.

## Z-TEST (DETAILED EXPLANATION)



1. What is Z-Test
The Z-test is an inferential statistical test used to determine whether a sample mean
is significantly different from a population mean when the population standard
deviation is known and/or the sample size is large.

It is based on the standard normal (Z) distribution.

---

2. When to Use Z-Test
Use a Z-test when:
- Population standard deviation (σ) is known
- Sample size is large (n ≥ 30)
- Data is quantitative
- Sample is randomly selected
- Population is normal OR sample size is large (Central Limit Theorem applies)

---

3. Types of Z-Tests

A. One-Sample Z-Test
Used to compare a sample mean with a known population mean.

Hypotheses:
H₀: μ = μ₀
H₁: μ ≠ μ₀  (two-tailed)
H₁: μ > μ₀  (right-tailed)
H₁: μ < μ₀  (left-tailed)

Test statistic:
Z = (x̄ − μ) / (σ / √n)

Where:
x̄ = sample mean
μ = population mean
σ = population standard deviation
n = sample size

---

B. Two-Sample Z-Test
Used to compare the means of two independent populations when both population
standard deviations are known.

Test statistic:
Z = (x̄₁ − x̄₂) / √(σ₁²/n₁ + σ₂²/n₂)

---

C. Z-Test for Proportion
Used to test a population proportion.

Test statistic:
Z = (p̂ − p) / √(pq / n)

Where:
p̂ = sample proportion
p = population proportion
q = 1 − p

---

4. Assumptions of Z-Test
- Random sampling
- Independence of observations
- Known population standard deviation
- Large sample size
- Normality of population or CLT applies

---

5. Decision Rule
At significance level α:

- Reject H₀ if |Zcalculated| > Zcritical
OR
- Reject H₀ if p-value < α

Common Z critical values:
- α = 0.05 → Z = ±1.96 (two-tailed)
- α = 0.01 → Z = ±2.58 (two-tailed)

---

6. Confidence Interval Using Z-Test
Confidence interval for population mean:

x̄ ± Z(α/2) · (σ / √n)

---

7. Example
Given:
Sample mean x̄ = 52
Population mean μ = 50
Population SD σ = 10
Sample size n = 100

Z = (52 − 50) / (10 / √100)
Z = 2

Since 2 > 1.96, reject H₀ at 5% level of significance.

---

8. Z-Test vs t-Test (Key Difference)

Z-Test:
- Population standard deviation known
- Large sample size
- Uses normal distribution

t-Test:
- Population standard deviation unknown
- Small sample size
- Uses t-distribution

---

9. Summary
The Z-test is used when the population standard deviation is known and the sample
size is large. The Z-test uses the standard normal distribution to test hypotheses
about population means and proportions.


## t-Test (Student’s t-Test)



1. Introduction
The t-test is an inferential statistical test used to determine whether there is a
statistically significant difference between means when:
- The sample size is small (usually n < 30), and/or
- The population standard deviation (σ) is unknown.

The t-test is based on the Student’s t-distribution, which is wider and flatter
than the normal distribution and depends on degrees of freedom.

---

2. When to Use a t-Test
Use a t-test when:
- Data are quantitative (interval or ratio scale)
- Samples are randomly selected
- Population variance is unknown
- Data are approximately normally distributed

---

3. Types of t-Tests

A. One-Sample t-Test
Used to compare the sample mean with a known or hypothesized population mean.

Hypotheses:
- H₀: μ = μ₀
- H₁: μ ≠ μ₀ (two-tailed)
       μ > μ₀ (right-tailed)
       μ < μ₀ (left-tailed)

Test statistic:
t = (x̄ − μ₀) / (s / √n)

Degrees of freedom:
df = n − 1

---

B. Independent (Two-Sample) t-Test
Used to compare the means of two independent samples.

Hypotheses:
- H₀: μ₁ = μ₂
- H₁: μ₁ ≠ μ₂ (or >, <)

Test statistic (equal variances assumed):
t = (x̄₁ − x̄₂) / √[ s_p² (1/n₁ + 1/n₂) ]

Where pooled variance:
s_p² = [ (n₁ − 1)s₁² + (n₂ − 1)s₂² ] / (n₁ + n₂ − 2)

Degrees of freedom:
df = n₁ + n₂ − 2

---

C. Paired Sample t-Test
Used when the two samples are related or matched
(e.g., before–after measurements on the same subjects).

Define:
d = difference between paired observations

Hypotheses:
- H₀: μ_d = 0
- H₁: μ_d ≠ 0

Test statistic:
t = d̄ / (s_d / √n)

Degrees of freedom:
df = n − 1

---

4. Assumptions of t-Test
- Random sampling
- Independence of observations
- Normality of data (especially for small samples)
- Equal variances (for independent t-test with pooled variance)

---

5. Decision Rule
At significance level α:
- Reject H₀ if |t_calculated| > t_critical
OR
- Reject H₀ if p-value < α

---

6. Confidence Interval Using t-Test
For population mean μ:
x̄ ± t(α/2, df) · (s / √n)

---

7. Advantages of t-Test
- Suitable for small samples
- Simple and widely used
- Robust to slight deviations from normality

---

 8. Limitations of t-Test
- Sensitive to outliers
- Assumes normality
- Not suitable for categorical data

---

9. Example
A sample of 10 students has:
- Mean score = 65
- Standard deviation = 5
Test whether the population mean score is 60 at α = 0.05.

Steps:
1. H₀: μ = 60, H₁: μ ≠ 60
2. Calculate t using the formula
3. Compare with t critical value
4. Draw conclusion

---

10. Summary
The t-test is a powerful inferential tool used to test hypotheses about population
means when the sample size is small and population variance is unknown.


## CHI-SQUARE TEST (χ² TEST)

1. What is Chi-Square Test
The Chi-Square (χ²) test is a non-parametric inferential statistical test used to
examine whether there is a significant difference between observed frequencies
and expected frequencies.

It is mainly used for categorical (qualitative) data.

---

2. When to Use Chi-Square Test
Use the Chi-Square test when:
- Data are categorical
- Data are in the form of frequencies
- Samples are randomly selected
- Observations are independent
- Expected frequency in each cell is at least 5 (rule of thumb)

---

3. Types of Chi-Square Tests

A. Chi-Square Test for Goodness of Fit
Used to check whether observed data fit a specified theoretical distribution.

Hypotheses:
H₀: Observed frequencies follow the expected distribution
H₁: Observed frequencies do not follow the expected distribution

Test statistic:
χ² = Σ ( (O − E)² / E )

Where:
O = observed frequency
E = expected frequency

Degrees of freedom:
df = k − 1
(k = number of categories)

---

B. Chi-Square Test of Independence
Used to determine whether two categorical variables are independent.

Hypotheses:
H₀: The two variables are independent
H₁: The two variables are dependent (associated)

Expected frequency:
E = (Row total × Column total) / Grand total

Test statistic:
χ² = Σ ( (O − E)² / E )

Degrees of freedom:
df = (r − 1)(c − 1)

Where:
r = number of rows
c = number of columns

---

C. Chi-Square Test of Homogeneity
Used to test whether different populations have the same distribution of a
categorical variable.

Hypotheses:
H₀: All populations have the same distribution
H₁: At least one population has a different distribution

Test statistic:
χ² = Σ ( (O − E)² / E )

Degrees of freedom:
df = (r − 1)(c − 1)

---

4. Assumptions of Chi-Square Test
- Random sampling
- Independence of observations
- Data in frequency form
- Expected frequencies ≥ 5

---

5. Decision Rule
At significance level α:

- Reject H₀ if χ²calculated > χ²critical
OR
- Reject H₀ if p-value < α

---

6. Example (Test of Independence)
A survey records gender and preference for a product.

Observed table:

Gender     Like   Dislike
Male       30     20
Female     20     30

Calculate expected frequencies using:
E = (Row total × Column total) / Grand total

Then compute χ² and compare with table value.

---

7. Advantages of Chi-Square Test
- Simple to apply
- No assumption of normality
- Suitable for categorical data

---

8. Limitations of Chi-Square Test
- Not suitable for small expected frequencies
- Does not measure strength of association
- Sensitive to sample size

---

9. Summary
The Chi-Square test is used to test relationships or differences involving
categorical data by comparing observed and expected frequencies.


## ANOVA (ANALYSIS OF VARIANCE)

1. What is ANOVA
ANOVA (Analysis of Variance) is an inferential statistical technique used to test
whether there are significant differences between the means of three or more
groups.

Instead of comparing means directly, ANOVA compares variances.

---

2. Why ANOVA is Used
- To compare more than two population means
- To control Type I error that would increase if multiple t-tests were used
- To determine whether at least one group mean differs from others

---

3. Types of ANOVA

A. One-Way ANOVA
Used when there is:
- One independent variable (factor)
- Three or more groups (levels)

Hypotheses:
H₀: μ₁ = μ₂ = μ₃ = ... = μₖ
H₁: At least one mean is different

---

B. Two-Way ANOVA
Used when there are:
- Two independent variables
- Tests main effects and interaction effects

Hypotheses:
H₀₁: No effect of factor A
H₀₂: No effect of factor B
H₀₃: No interaction between factors A and B

---

4. Assumptions of ANOVA
- Random sampling
- Independence of observations
- Normality of populations
- Homogeneity of variances (equal variances)

---

5. ANOVA Terminology
- Factor: Independent variable
- Levels: Categories of a factor
- Treatment: Specific combination of factor levels
- Response variable: Dependent variable

---

6. ANOVA Table (One-Way)

Source of Variation | Sum of Squares | Degrees of Freedom | Mean Square | F-ratio
Between Groups      | SSB            | k − 1              | MSB         | MSB / MSW
Within Groups       | SSW            | N − k              | MSW         |
Total               | SST            | N − 1              |             |

Where:
SSB = Sum of squares between groups
SSW = Sum of squares within groups
SST = Total sum of squares

---

7. F-Test in ANOVA
Test statistic:
F = MSB / MSW

Decision Rule:
- Reject H₀ if Fcalculated > Fcritical
OR
- Reject H₀ if p-value < α

---

8. Example
Three teaching methods are used to teach students.
Their test scores are compared using one-way ANOVA.

H₀: All teaching methods produce the same mean score
H₁: At least one method produces a different mean score

Calculate F and compare with table value.

---

9. Advantages of ANOVA
- Compares multiple means at once
- Controls Type I error
- Efficient and widely used

---

10. Limitations of ANOVA
- Does not indicate which groups differ
- Sensitive to assumption violations
- Requires post-hoc tests for detailed comparison

---

11. Post-Hoc Tests
Used after rejecting H₀ to find which means differ:
- Tukey’s HSD
- Bonferroni test
- Scheffé test

---

12. Summary
ANOVA tests whether there are statistically significant differences among group
means by analyzing variance and using the F-test.
