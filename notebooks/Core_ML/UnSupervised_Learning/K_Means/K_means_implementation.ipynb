## K-Means Clustering

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
from sklearn.datasets import make_blobs
```

```python
X,y = make_blobs(
    centers=3,
    n_features=2,
    random_state=42,
    n_samples=1000,
)
```

```python
X
```

```python
y
```

Not taking y as we going unsupervised

```python
plt.scatter(X[:,0], X[:,1], c=y)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
# scaler.fit_transform()
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=42)
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.cluster import KMeans
```

```python
# ELbo Method to select K Value
wcss=[]
for k in range(1,30):
    km = KMeans(n_clusters=k, init = 'k-means++')
    km.fit(X_train_scaled)
    wcss.append(km.inertia_)
```

```python
wcss
```

```python
# plot elbow curve
plt.plot(range(0,len(wcss)),wcss)
plt.xticks(range(1,len(wcss)))
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show()
```

```python
# ELbo Method to select K Value
wcss=[]
for k in range(1,15):
    km = KMeans(n_clusters=k, init = 'k-means++')
    km.fit(X_train_scaled)
    wcss.append(km.inertia_)
wcss
# plot elbow curve
plt.plot(range(0,len(wcss)),wcss)
plt.xticks(range(1,len(wcss)))
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show()
```

```python
# ELbo Method to select K Value
wcss=[]
for k in range(1,7):
    km = KMeans(n_clusters=k, init = 'k-means++')
    km.fit(X_train_scaled)
    wcss.append(km.inertia_)
wcss
# plot elbow curve
plt.plot(range(0,len(wcss)),wcss)
plt.xticks(range(1,len(wcss)))
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show()
```

we see the elbow is at 2!

```python
from sklearn.cluster import KMeans

for k in [2, 3]:
    km = KMeans(n_clusters=k, init='k-means++', random_state=42)
    y_km = km.fit_predict(X_train_scaled)
    plt.figure(figsize=(6,4))
    plt.scatter(X_train_scaled[:,0], X_train_scaled[:,1], c=y_km, cmap='viridis')
    plt.title(f"KMeans with k={k}")
    plt.show()
```

```python
from sklearn.metrics import silhouette_score

for k in range(2, 7):
    km = KMeans(n_clusters=k, random_state=42)
    y_km = km.fit_predict(X_train_scaled)
    score = silhouette_score(X_train_scaled, y_km)
    print(f"k={k}, silhouette={score:.3f}")
```

```python
## Validating the k_value:
### Knee locator
### Silhouette Scoring
```

```python
wcss=[]
for k in range(1,11):
    km = KMeans(n_clusters=k, init = 'k-means++')
    km.fit(X_train_scaled)
    wcss.append(km.inertia_)
wcss
```

```python
from kneed import KneeLocator

kl = KneeLocator(range(1 ,11),wcss,curve= "convex",direction="decreasing")
```

```python
kl.elbow
```

```python
# Silhouette scoring
from sklearn .metrics import silhouette_score
```

```python
silhouette_coef = []
for k in range(2,11):
    kmeans=KMeans(n_clusters=k,init = 'k-means++')
    kmeans.fit(X_train_scaled)
    score = silhouette_score(X_train_scaled,kmeans.labels_)
    silhouette_coef.append(score)
```

```python
silhouette_coef
```

```python
plt.plot(range(2,11),silhouette_coef)
plt.xticks(range(2,11))
plt.xlabel("Number of Clusters")
plt.ylabel("silhouette_coef")
plt.show()
```
