# Hierarchial Clustering

*Testing HC on iris dataset to see how many clusters it proposes with PCA and different Distance Metrics:*

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
# loading dataset
from sklearn.datasets import load_iris
iris = load_iris()
````

````python
iris
````

````python
print(iris.DESCR)
````

````python
iris_data = pd.DataFrame(iris.data)
iris_data
````

````python
# applying Standardization and PCA
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
````

````python
iris_scaled = scaler.fit_transform(iris_data)
iris_scaled
````

````python
# now PCA:
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
````

````python
iris_scaled_pca = pca.fit_transform(iris_scaled)
iris_scaled_pca
````

````python
plt.scatter(iris_scaled_pca[:,0],iris_scaled_pca[:,1])
````

````python
# cheating to see ho many clusters we actually got just to compare:
plt.scatter(iris_scaled_pca[:,0],iris_scaled_pca[:,1],c=iris.target)
````

````python
# now making the dendrogram
import scipy.cluster.hierarchy as sc
plt.figure(figsize=(20,10))
plt.title("Dendrogram")

sc.dendrogram(sc.linkage(iris_scaled_pca))
````

````python
# now making the dendrogram
import scipy.cluster.hierarchy as sc
plt.figure(figsize=(20,10))
plt.title("Dendrogram")

sc.dendrogram(sc.linkage(iris_scaled_pca,method='ward'))
plt.title('Dendogram_ward')
plt.xlabel('Sample Index')
plt.ylabel('Eucledian Distance')
````

````python
plt.figure(figsize=(20,10))
plt.title("Dendrogram")

sc.dendrogram(sc.linkage(iris_scaled_pca,method='single'))
plt.title('Dendogram_single')
plt.xlabel('Sample Index')
plt.ylabel('Eucledian Distance')
````

````python
plt.figure(figsize=(20,10))
plt.title("Dendrogram")

sc.dendrogram(sc.linkage(iris_scaled_pca,method='complete'))
plt.title('Dendogram_complete')
plt.xlabel('Sample Index')
plt.ylabel('Eucledian Distance')
````

Let’s interpret your results carefully instead of just reading numbers.

You’re working on the Iris dataset, which has 3 biological classes — but clustering doesn’t know that.

---

# 🔎 1️⃣ Why Is k = 2 Best by Silhouette?

You got:

```
k=2 → 0.577  (highest)
k=3 → 0.447
k=4 → 0.401
...
```

This is very meaningful.

Silhouette measures:

[
s(i) = \frac{b(i) - a(i)}{\max(a(i), b(i))}
]

Where:

* (a(i)) = average distance within cluster
* (b(i)) = distance to nearest other cluster

Large value = well-separated clusters.

---

## What’s happening geometrically?

In Iris:

* Setosa is **very far** from the other two.
* Versicolor & Virginica **overlap significantly**.

So the natural geometric structure is:

> 1 very clean cluster
> 1 mixed cluster

So k=2 gives:

* Cluster 1 → Setosa
* Cluster 2 → Other two species mixed

That separation is strong → silhouette high (0.577).

When you force k=3:

* The algorithm tries to split overlapping species
* Boundaries become fuzzy
* Silhouette drops

This is a deep insight:

> Clustering reveals geometric structure, not biological labels.

The data geometry prefers 2 clusters.

Biology says 3 species.

These are not always aligned.

---

# 🔎 2️⃣ Why Did Single Linkage Score Highest for k=3?

You got:

```
single  → 0.504
average → 0.480
complete→ 0.449
ward    → 0.446
```

That’s interesting — and slightly surprising.

Normally on Iris:

* Ward performs best in ARI (agreement with labels)
* Single linkage suffers from chaining

But silhouette measures *compactness and separation*, not agreement with truth.

Single linkage tends to:

* Create one tight cluster (setosa)
* Leave a stretched cluster for others
* Which can inflate silhouette

So silhouette ≠ biological correctness.

---

# 🔎 3️⃣ Adjusted Rand Index ≈ 0.558

You computed:

```
ARI = 0.558
```

ARI compares clustering with true labels.

Range:

* 1.0 → perfect match
* 0 → random
* <0 → worse than random

0.56 means:

> Moderate agreement, not great.

This is expected because:

* Versicolor & Virginica overlap.
* No unsupervised method can perfectly separate them in original feature space.

Even k-means clustering usually gives ARI around 0.6–0.7 on Iris.

So your result is reasonable.

---

# 🧠 Big Conceptual Lessons From Your Experiment

### 1️⃣ Clustering does NOT recover labels automatically

Even on “clean” datasets like Iris.

---

### 2️⃣ Silhouette does NOT necessarily choose true class count

It chooses geometric separability.

Here:

* Geometry suggests k=2.
* Labels say k=3.

Both are “correct” in different senses.

---

### 3️⃣ Different Linkages Optimize Different Geometry

Ward:

* Prefers spherical clusters.
* Often aligns better with labels.

Single:

* Allows irregular shapes.
* Can inflate silhouette.
