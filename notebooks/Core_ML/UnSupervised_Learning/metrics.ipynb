## Top to bottom of maths and metrics for unsupervised

In unsupervised learning, “metrics” can mean three closely related but different things:

1. **Distance / similarity measures** (what defines closeness in the space)
2. **Objective functions** (what the algorithm optimizes)
3. **Evaluation metrics** (how we judge clustering or structure quality)

Below is a **top-to-bottom view**: from mathematical foundations → algorithm objectives → practical evaluation.

---

# 1️⃣ Theoretical Foundations: What Is a Metric?

## Mathematical Metric (Distance Function)

A function ( d(x, y) ) is a **metric** if it satisfies:

1. Non-negativity: ( d(x,y) \ge 0 )
2. Identity: ( d(x,y)=0 \iff x=y )
3. Symmetry: ( d(x,y)=d(y,x) )
4. Triangle inequality: ( d(x,z) \le d(x,y)+d(y,z) )

Not all similarity measures are true metrics.

---

## Common Distance Functions

### Euclidean Distance

[
d(x,y) = \sqrt{\sum (x_i - y_i)^2}
]

* Assumes isotropic structure
* Used in **K-means clustering**

---

### Manhattan Distance (L1)

[
\sum |x_i - y_i|
]

* More robust to outliers
* Works better in high dimensions

---

### Cosine Distance

[
1 - \frac{x \cdot y}{||x|| ||y||}
]

* Used in text embeddings
* Not a proper metric (triangle inequality may fail)

---

### Mahalanobis Distance

[
\sqrt{(x-\mu)^T \Sigma^{-1} (x-\mu)}
]

* Accounts for covariance structure
* Used in **Gaussian Mixture Model**

---

### Kernel-Induced Distance

Used in spectral methods such as **Spectral clustering**

Transforms space implicitly via:
[
K(x,y) = \phi(x)^T \phi(y)
]

---

# 2️⃣ Algorithm-Level Objective Functions

Unsupervised algorithms optimize internal criteria.

---

## A) Clustering Objectives

### K-Means Objective

Minimize within-cluster variance:

[
\sum_{k=1}^{K} \sum_{x \in C_k} ||x - \mu_k||^2
]

Optimizes compactness.

---

### Gaussian Mixture Model (Likelihood)

Maximizes log-likelihood:

[
\sum \log \left( \sum_k \pi_k \mathcal{N}(x | \mu_k, \Sigma_k) \right)
]

Probabilistic clustering.

---

### DBSCAN Density Criterion

Used in **DBSCAN**

Core ideas:

* ε-neighborhood
* Minimum points
* Density reachability

No global objective — density-based structure.

---

### Spectral Clustering Objective

Minimizes graph cut:

* Ratio cut
* Normalized cut

Based on graph Laplacian eigenvalues.

---

# 3️⃣ Internal Evaluation Metrics (No Ground Truth)

Used when labels are unknown.

---

## Silhouette Score

[
s = \frac{b-a}{\max(a,b)}
]

* ( a ): intra-cluster distance
* ( b ): nearest-cluster distance

Range: [-1, 1]

Good for compact + separated clusters.

---

## Davies–Bouldin Index

[
\frac{1}{K} \sum_k \max_{j \neq k} \frac{S_k + S_j}{M_{kj}}
]

Lower is better.

Measures cluster overlap.

---

## Calinski–Harabasz Index

[
\frac{\text{Between-cluster dispersion}}{\text{Within-cluster dispersion}}
]

Higher is better.

---

## Inertia (K-Means)

Within-cluster sum of squares.

Monotonically decreases with K — not reliable alone.

---

# 4️⃣ External Evaluation Metrics (If Labels Exist)

Used for benchmarking.

---

## Adjusted Rand Index (ARI)

Measures similarity between two clusterings.

Corrected for chance.

---

## Normalized Mutual Information (NMI)

Information-theoretic measure:

[
\frac{I(X;Y)}{\sqrt{H(X)H(Y)}}
]

Captures shared information.

---

## Fowlkes–Mallows Score

Precision-recall based clustering similarity.

---

# 5️⃣ Density & Topology-Based Metrics

Used in advanced unsupervised learning.

---

## Reachability Distance

Used in OPTICS.

Measures density connectivity.

---

## Persistent Homology

Topological data analysis:

* Captures shape across scales
* Uses Betti numbers

---

# 6️⃣ Representation Learning Metrics

Used in autoencoders and contrastive learning.

---

## Reconstruction Error

[
||x - \hat{x}||^2
]

Used in autoencoders.

---

## Contrastive Loss (InfoNCE)

Used in representation learning:
[
-\log \frac{e^{sim(x_i,x_j)/\tau}}{\sum e^{sim(x_i,x_k)/\tau}}
]

## Silhouette Score (Clustering Evaluation)

The **Silhouette Score** measures how well a data point fits within its cluster compared to other clusters. It evaluates **cluster cohesion vs. cluster separation** without requiring ground-truth labels.

---

# 1️⃣ Silhouette Formula

For a data point ( i ):

[
s(i) = \frac{b(i) - a(i)}{\max(a(i), b(i))}
]

Where:

* (a(i)) = **average distance between point (i) and all other points in the same cluster**
* (b(i)) = **minimum average distance between point (i) and points in another cluster**

---

# 2️⃣ Formula for (a(i)) — Intra-Cluster Distance

Let (C_i) be the cluster containing point (i), and (|C_i|) its size.

[
a(i) =
\frac{1}{|C_i| - 1}
\sum_{j \in C_i,, j \ne i}
d(i, j)
]

Where:

* (d(i,j)) = distance between points (i) and (j)

Meaning:

Average distance from point (i) to **all other points in its cluster**.

Small (a(i)) → cluster is **compact**.

---

# 3️⃣ Formula for (b(i)) — Nearest Cluster Distance

Let (C_k) be another cluster where (k \ne i).

First compute average distance to cluster (C_k):

[
d(i, C_k) =
\frac{1}{|C_k|}
\sum_{j \in C_k}
d(i, j)
]

Then choose the **nearest cluster**:

[
b(i) = \min_{k \ne i} d(i, C_k)
]

Meaning:

Average distance from point (i) to the **closest neighboring cluster**.

Large (b(i)) → clusters are **well separated**.

---

# 4️⃣ Final Silhouette Value

[
s(i) = \frac{b(i) - a(i)}{\max(a(i), b(i))}
]

Interpretation:

| Value              | Meaning                          |
| ------------------ | -------------------------------- |
| ( s(i) \approx 1 ) | Point clearly inside its cluster |
| ( s(i) \approx 0 ) | Point lies between clusters      |
| ( s(i) < 0 )       | Likely misclassified             |

---

# 5️⃣ Dataset Silhouette Score

For (n) data points:

[
S =
\frac{1}{n}
\sum_{i=1}^{n} s(i)
]

This gives the **overall clustering quality**.

Higher score → better cluster structure.

---

# 6️⃣ Key Idea

Silhouette balances two forces:

**Cohesion (within cluster)**

[
a(i)
]

**Separation (between clusters)**

[
b(i)
]

Good clustering satisfies:

[
a(i) \ll b(i)
]

So:

[
s(i) \rightarrow 1
]

# 1️⃣ Why Silhouette Score Fails in High Dimensions

Silhouette for point ( i ):

[
s(i) = \frac{b(i) - a(i)}{\max(a(i), b(i))}
]

Where:

* ( a(i) ) = mean intra-cluster distance
* ( b(i) ) = mean nearest-cluster distance

---

## The Core Problem: Distance Concentration

In high dimensions:

[
\frac{d_{max} - d_{min}}{d_{min}} \to 0
]

Distances become almost equal.

Let’s see why.

Assume:

* Data sampled from ( \mathcal{N}(0, I_d) )
* Euclidean distance between two random points

Expected squared distance:

[
E[||x-y||^2] = 2d
]

Variance:

[
Var(||x-y||^2) = 8d
]

Standard deviation scales as ( \sqrt{d} )

But mean scales as ( d )

So:

[
\frac{\text{Std}}{\text{Mean}} \sim \frac{\sqrt{d}}{d} = \frac{1}{\sqrt{d}} \to 0
]

Distances collapse toward a constant.

---

## Effect on Silhouette

If:

* ( a(i) \approx c )
* ( b(i) \approx c + \epsilon )

Then:

[
s(i) \approx \frac{\epsilon}{c}
]

As dimension increases:

[
\epsilon \to 0
]

So silhouette → 0

Even when clusters are real.

---

## Production Implication

* Silhouette underestimates clustering quality in embeddings (e.g., 768-dim transformer vectors)
* It falsely suggests “no structure”
* Bad metric for high-dim text / vision embeddings

Better alternatives in production:

* Stability under resampling
* Downstream task lift
* Density-based validation
* Visualization in reduced space (UMAP)

# 2️⃣ Spectral vs Density Clustering (Mathematical Comparison)

We compare:

* Spectral clustering
* DBSCAN

---

## Spectral Clustering (Graph Theory Based)

Construct similarity graph:

[
W_{ij} = \exp(-||x_i-x_j||^2 / \sigma^2)
]

Compute Laplacian:

[
L = D - W
]

Minimize:

[
\text{Tr}(H^T L H)
]

Subject to orthogonality.

Interpretation:

* Minimizes graph cut
* Finds low-conductance partitions
* Solves relaxed combinatorial problem via eigenvectors

---

### Geometric Meaning

It finds directions where:

Graph connectivity is weakest.

Equivalent to:

Embedding data into eigenspace where clusters become separable.

---

## Density Clustering (DBSCAN)

Defines cluster as:

Connected component of points where:

[
|N_\epsilon(x)| \geq minPts
]

No global optimization.

Clusters = maximal density-connected sets.

---

## Mathematical Contrast

| Aspect       | Spectral            | Density                  |
| ------------ | ------------------- | ------------------------ |
| Structure    | Graph partition     | Level-set of density     |
| Optimization | Eigen-decomposition | None (local expansion)   |
| Shape        | Handles non-convex  | Handles arbitrary shapes |
| Sensitive to | Similarity kernel   | ε scale                  |
| Complexity   | O(n³) naive         | O(n log n) with index    |

---

## Production View

Use Spectral when:

* Graph structure matters (social, recommender, similarity graphs)
* Clear connectivity pattern
* Small-to-medium datasets

Use DBSCAN when:

* Noise matters
* Arbitrary shape
* Outlier detection needed
* You want no k parameter

# 4️⃣ Deep Geometric Interpretation of Clustering

Clustering = partitioning space based on geometry of data manifold.

Three geometric views:

---

## 1) Voronoi Partition (KMeans)

Space divided into convex polytopes.

Assumes:

* Clusters are convex
* Equal variance
* Spherical geometry

---

## 2) Density Level Sets

Cluster = connected component of:

[
{x : p(x) > \lambda}
]

Used implicitly in DBSCAN.

More flexible.

---

## 3) Graph Conductance

Cluster = subset minimizing:

[
\frac{cut(A, B)}{\min(vol(A), vol(B))}
]

Used in spectral.

Detects bottlenecks.

---

## Production Insight

Different algorithms assume different geometry:

| Algorithm | Assumption          |
| --------- | ------------------- |
| KMeans    | Spherical convex    |
| GMM       | Elliptical Gaussian |
| DBSCAN    | Density-connected   |
| Spectral  | Weak graph cut      |

Choosing algorithm = choosing geometry.

# 3️⃣ Formal Cluster Stability Theory

Now the mathematical part.

---

## Definition: Algorithmic Stability

Let:

[
C(D)
]

Be clustering on dataset D.

We say clustering is stable if:

[
C(D) \approx C(D')
]

When D' is small perturbation of D.

---

## Perturbation Model

Let D' be:

* Bootstrap sample
* Noisy version
* Subsample

Define similarity measure between clusterings:

Adjusted Rand Index (ARI).

Stability score:

[
S = E_{D'}[ARI(C(D), C(D'))]
]

---

## Theoretical Insight

Ben-David et al (2006) showed:

If clustering objective has:

* Unique global optimum
* Large margin between best and second-best solution

Then clustering is stable.

---

## Stability–Bias Tradeoff

* Very large K → unstable
* Very small K → stable but high bias

Stability often peaks at “correct” K.

---

## Margin Interpretation

For KMeans:

If clusters are well-separated:

[
\min_{i \neq j} ||\mu_i - \mu_j|| >> \max_k \sigma_k
]

Then:

Perturbations unlikely to swap assignments.

Thus high stability.

---

## Practical Stability Protocol

1. Sample 80% data
2. Cluster
3. Repeat 20 times
4. Compute ARI matrix
5. Choose hyperparameters with max mean ARI

Much more reliable than silhouette.
