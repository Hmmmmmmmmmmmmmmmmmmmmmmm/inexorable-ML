## PCA

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
```

```python
cancer_dataset = load_breast_cancer()
```

```python
cancer_dataset
```

```python
cancer_dataset.keys()
```

```python
print(cancer_dataset.DESCR)
```

```python
cancer_dataset.data
```

```python
df = pd.DataFrame(cancer_dataset['data'],columns=cancer_dataset['feature_names'])
df
```

```python
null_counts = df.isnull().sum()
null_counts = null_counts[null_counts > 0]

if not null_counts.empty:
    print(null_counts)
```

```python
df.info()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
scaler.fit(df)
```

```python
scaled_df = scaler.transform(df)
```

```python
scaled_df
```

```python
from sklearn.decomposition import PCA
```

```python
pca3 = PCA(n_components=3)
pca2 = PCA(n_components=2)
```

```python
data_pca2 = pca2.fit_transform(scaled_df)
data_pca3 = pca3.fit_transform(scaled_df)
```

```python
data_pca2
```

```python
pca2.components_
```

```python
pca2.explained_variance_
```

```python
data_pca3
```

```python
plt.figure(figsize=(8,6))
plt.scatter(data_pca2[:,0],data_pca2[:,1], c = cancer_dataset['target'],cmap = 'cool')
plt.xlabel("First principal component")
plt.ylabel("Second Principal Component ")
```

```python
from mpl_toolkits.mplot3d import Axes3D
 # not always required in newer versions

fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(
    data_pca3[:,0],
    data_pca3[:,1],
    data_pca3[:,2],
    c=cancer_dataset['target'],
    cmap='cool'
)

ax.set_xlabel("First Principal Component")
ax.set_ylabel("Second Principal Component")
ax.set_zlabel("Third Principal Component")

plt.show()
```

```python
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

scatter = ax.scatter(
    data_pca3[:,0],
    data_pca3[:,1],
    data_pca3[:,2],
    c=cancer_dataset['target'],
    cmap='cool'
)

ax.set_xlabel("First Principal Component")
ax.set_ylabel("Second Principal Component")
ax.set_zlabel("Third Principal Component")

ax.view_init(elev=30, azim=190)  # try different numbers

plt.show()
```

```python
import plotly.express as px

df_plot = pd.DataFrame({
    "PC1": data_pca3[:,0],
    "PC2": data_pca3[:,1],
    "PC3": data_pca3[:,2],
    "target": cancer_dataset['target']
})

fig = px.scatter_3d(
    df_plot,
    x="PC1",
    y="PC2",
    z="PC3",
    color="target",
    size_max=1
)

fig.show()
```
