# 1. What Anomaly Detection Is (Core Theory)

**Anomaly Detection** = identifying **data points that significantly deviate from normal patterns**.

Synonyms you should know:

* Outlier detection
* Novelty detection
* Rare event detection
* Fraud detection

### Types of anomalies

1. **Point anomaly**

   * A single data point is abnormal
   * Example: credit card transaction of ₹1,00,000 from a normally ₹500 account.

2. **Contextual anomaly**

   * Depends on context (time/location)
   * Example: 30°C temperature is normal but abnormal in winter.

3. **Collective anomaly**

   * A group of points is abnormal
   * Example: unusual sequence in network traffic.

[

[

You must understand:

[

]

Important for **elliptical distributions**.

# 1. Isolation Forest Anomaly Detection

Implemented in libraries like scikit-learn.

---

# Intuition

Most algorithms detect anomalies by **modeling normal points**.

Isolation Forest does the opposite:

👉 **Anomalies are easier to isolate than normal points.**

Example:

Imagine random splits in data:

Normal points:

* clustered together
* require many splits to isolate

Anomalies:

* isolated
* require fewer splits

Thus:

**shorter path length → anomaly**

---

# Mathematical Idea

Isolation trees randomly partition feature space.

At each node:

Randomly choose

* feature
* split value

Average path length for point (x):

[
h(x)
]

Expected path length:

[
E(h(x))
]

Anomaly score:

[
s(x,n)=2^{-\frac{E(h(x))}{c(n)}}
]

Where:

* (n) = sample size
* (c(n)) = normalization constant

Interpretation:

| Score | Meaning     |
| ----- | ----------- |
| ~1    | anomaly     |
| ~0.5  | normal      |
| <0.5  | very normal |

---

# Algorithm Steps

1. Randomly sample dataset.
2. Build many **Isolation Trees**.
3. For each tree:

   * randomly select feature
   * randomly select split value.
4. Measure **path length to isolate point**.
5. Average path length across trees.
6. Compute anomaly score.

---

# Complexity

Training:

[
O(t \cdot \psi \log \psi)
]

Where:

* (t) = number of trees
* (\psi) = subsample size.

Very scalable.

---

# Python Implementation

Using scikit-learn

```python
from sklearn.ensemble import IsolationForest

model = IsolationForest(
    n_estimators=100,
    contamination=0.05,
    random_state=42
)

model.fit(X)

predictions = model.predict(X)

scores = model.decision_function(X)
```

✖ random splits sometimes unstable

# 2. DBSCAN Clustering Anomaly Detection

DBSCAN = **Density-Based Spatial Clustering of Applications with Noise**

Available in scikit-learn.

---

# Intuition

Clusters are **dense regions**.

Points that do not belong to any cluster are **anomalies**.

Thus:

dense region → normal
sparse region → anomaly

---

# Important Parameters

### eps (ε)

Neighborhood radius.

### min_samples

Minimum points needed to form cluster.

---

# Types of Points

### Core point

Has ≥ min_samples neighbors.

---

### Border point

Neighbor of core point but not dense.

---

### Noise point

Not reachable from any cluster.

👉 **Noise points = anomalies**

---

# Mathematical Idea

Neighborhood of point:

[
N_{\epsilon}(p) = {q \mid dist(p,q) \le \epsilon}
]

Core condition:

[
|N_{\epsilon}(p)| \ge MinPts
]

---

# Algorithm Steps

1. Pick random point.
2. Find neighbors within ε.
3. If neighbors ≥ min_samples → create cluster.
4. Expand cluster.
5. Points not belonging to cluster → noise.

---

# Python Implementation

Using scikit-learn.

```python
from sklearn.cluster import DBSCAN

model = DBSCAN(eps=0.5, min_samples=5)

labels = model.fit_predict(X)
```

Label values:

| Label | Meaning    |
| ----- | ---------- |
| -1    | anomaly    |
| ≥0    | cluster ID |

---

# Strengths

✔ Detects **arbitrary cluster shapes**
✔ Automatically finds anomalies
✔ No need to specify number of clusters

---

# Weaknesses

✖ sensitive to eps
✖ struggles in high dimensions
✖ uneven densities problem

---

# When to Use

* spatial data
* geographic anomaly detection
* customer segmentation anomalies

# 3. Local Outlier Factor (LOF)

Also available in scikit-learn.

---

# Intuition

LOF compares **local density of a point vs its neighbors**.

If a point has **much lower density than neighbors → anomaly**.

Example:

Cluster of dense points with one sparse point inside → anomaly.

---

# Core Idea

Instead of global density:

LOF uses **local density**.

---

# Mathematical Components

### k-distance

Distance to kth nearest neighbor.

[
k\text{-distance}(p)
]

---

### Reachability distance

[
reach_dist_k(p,o) = max(k\text{-distance}(o), dist(p,o))
]

---

### Local Reachability Density (LRD)

[
LRD(p)=\frac{1}{\frac{\sum reach_dist_k(p,o)}{|N_k(p)|}}
]

---

### LOF score

[
LOF(p) =
\frac{\sum \frac{LRD(o)}{LRD(p)}}{|N_k(p)|}
]

Interpretation:

| LOF | Meaning        |
| --- | -------------- |
| ≈1  | normal         |
| >1  | anomaly        |
| >>1 | strong anomaly |

---

# Algorithm Steps

1. Compute k-nearest neighbors.
2. Compute reachability distance.
3. Compute local density.
4. Compare density with neighbors.

---

# Python Implementation

Using scikit-learn.

```python
from sklearn.neighbors import LocalOutlierFactor

model = LocalOutlierFactor(
    n_neighbors=20,
    contamination=0.05
)

pred = model.fit_predict(X)
```

Output:

* **-1 → anomaly**
* **1 → normal**

---

# Strengths

✔ detects **local anomalies**
✔ good for uneven densities

---

# Weaknesses

✖ computationally expensive
✖ not scalable to very large datasets

Ah! Now I see exactly what you’re asking about — the **expected path length (E(h(x))) in Isolation Forest**. Let’s go **carefully step by step**, because the notation can be confusing.

---

# 1️⃣ What “Expected” Means

In probability theory:

[
E[X] = \text{expected value of a random variable } X
]

* Intuition: it’s the **average value you’d get if you repeated the random experiment many times**.
* Formula for a discrete random variable:

[
E[X] = \sum_i x_i , P(X=x_i)
]

For a continuous variable:

[
E[X] = \int x , f_X(x) dx
]

So yes, **“E()” is basically an average**, weighted by probability.

---

# 2️⃣ In Isolation Forest Context

Isolation Forest is a **randomized algorithm**. Each tree is built by:

1. Randomly selecting a feature
2. Randomly selecting a split value

Now, the **path length (h(x))** is the **number of splits it takes to isolate point x in a tree**.

Because trees are random:

* (h(x)) is **not fixed** — it changes depending on the random splits.
* But we can **average it across many trees** → that’s exactly the **expected path length**.

So:

[
E(h(x)) = \text{average path length of x across all random trees}
]

---

# 3️⃣ Why It Works

* Anomalies are usually **isolated quickly** → small (h(x)) → high anomaly score
* Normal points are **harder to isolate** → large (h(x)) → low anomaly score

We normalize (E(h(x))) by (c(n)) to make it comparable across sample sizes:

[
s(x,n) = 2^{-\frac{E(h(x))}{c(n)}}
]

Where:

* (c(n) \approx 2H(n-1) - \frac{2(n-1)}{n})
* (H(i)) = harmonic number (≈ ln(i) + 0.5772)

---

# ✅ TL;DR

1. (E()) = expected value = **average** over all random trees
2. (h(x)) = path length in a single random tree
3. (E(h(x))) = **average path length** across many Isolation Trees
4. Small (E(h(x))) → anomaly, large → normal

So it’s **not probability**, it’s an **average over random experiments** (the “random” trees).
