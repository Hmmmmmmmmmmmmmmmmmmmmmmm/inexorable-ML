# AdaBoost


---

# 1️⃣ What is Boosting?

**Boosting** is an ensemble learning technique that:

* Combines many **weak learners**
* Trains them **sequentially**
* Each new model focuses on correcting the mistakes of the previous ones
* Final prediction = **weighted combination of all models**

### Key Idea:

Instead of training models independently (like Random Forest), boosting trains them **one after another**, each trying to fix errors made before.

---

# 2️⃣ What is a Weak Learner?

A weak learner is a model that:

* Performs slightly better than random guessing
* Is usually very simple

Most common choice:

* **Decision stump** (a decision tree with depth = 1)

Example stump:

```
if Age > 30 → Predict Yes
else → Predict No
```

Alone → weak
Combined sequentially → powerful

---

# 3️⃣ AdaBoost (Adaptive Boosting)

AdaBoost was introduced in 1996 by Freund & Schapire.

It was the first successful boosting algorithm.

The name comes from:

> Adaptive Boosting
> Because it **adapts** by focusing more on previously misclassified samples.

---

# 4️⃣ Core Idea of AdaBoost

Instead of changing the dataset,
AdaBoost changes the **weights of the training samples**.

Misclassified samples → get higher weight
Correctly classified samples → get lower weight

Next model focuses more on hard examples.

---

# 5️⃣ AdaBoost Algorithm (Binary Classification)

Let’s assume:

* Labels are **+1 and -1**
* Weak learner = decision stump

---

## Step 1: Initialize Weights

All training samples start equally weighted:

[
w_i = \frac{1}{N}
]

---

## Step 2: Train Weak Learner

Train model ( h_1(x) ) using weighted data.

---

## Step 3: Compute Weighted Error

[
\text{Error} = \sum w_i \cdot I(y_i \neq h(x_i))
]

This is NOT regular accuracy — it is **weighted error**.

---

## Step 4: Compute Model Weight

Each weak learner gets a weight:

[
\alpha = \frac{1}{2} \ln \left(\frac{1 - error}{error}\right)
]

Interpretation:

* Small error → large α → strong influence
* Error = 0.5 → α = 0 → useless learner
* Error > 0.5 → worse than random

---

## Step 5: Update Sample Weights

If correctly classified:
[
w_i \leftarrow w_i e^{-\alpha}
]

If misclassified:
[
w_i \leftarrow w_i e^{\alpha}
]

Then normalize weights so they sum to 1.

---

## Step 6: Repeat

Repeat for T rounds.

---

# 6️⃣ Final Prediction

Final model:

[
F(x) = \sum_{t=1}^{T} \alpha_t h_t(x)
]

Final class:

[
\text{sign}(F(x))
]

So it's a **weighted vote** of weak learners.

---

# 7️⃣ Intuition Behind the Formula

Why this strange log formula?

Because AdaBoost minimizes **exponential loss**:

[
L = e^{-yF(x)}
]

It’s mathematically elegant and leads to that α formula.

We’ll connect this to Gradient Boosting later — very important.

---

# 8️⃣ Why AdaBoost Works So Well

It:

* Focuses on hard examples
* Builds strong classifier from weak ones
* Automatically selects important features (via stumps)
* Often doesn’t overfit easily (surprisingly)

---

# 9️⃣ Strengths of AdaBoost

✅ Simple
✅ Theoretically elegant
✅ Works well with clean data
✅ No heavy hyperparameter tuning

---

# 🔟 Weaknesses

❌ Sensitive to noisy data
❌ Sensitive to outliers (because misclassified points get huge weights)
❌ Not great for large datasets compared to modern methods

---

# 1️⃣1️⃣ Geometric View (Important)

AdaBoost builds an **additive model**:

[
F(x) = \sum \alpha_t h_t(x)
]

Each stump adds a new dimension in function space.

Later you'll see:

> Gradient Boosting generalizes this idea using gradient descent.

---

# Summary of AdaBoost

| Step  | What Happens                     |
| ----- | -------------------------------- |
| 1     | Initialize equal weights         |
| 2     | Train weak learner               |
| 3     | Compute weighted error           |
| 4     | Compute learner weight α         |
| 5     | Increase weight of misclassified |
| 6     | Repeat                           |
| Final | Weighted vote                    |
