# XGBoost

## What is **XGBoost** in Machine Learning?

**XGBoost (Extreme Gradient Boosting)** is a highly efficient and scalable implementation of **gradient boosted decision trees** designed for supervised learning problems (regression and classification).

It was developed by Tianqi Chen and became widely popular after winning many Kaggle competitions.

---

# 1️⃣ Core Idea (Intuition)

XGBoost is based on **boosting**, specifically **gradient boosting**.

### Boosting Concept:

* Combine many **weak learners** (usually shallow decision trees)
* Each new tree corrects the errors of previous trees
* Final prediction = sum of all tree outputs

Instead of building one strong model, it builds many small trees sequentially.

---

# 2️⃣ Algorithm Overview (Step-by-Step)

Assume dataset:

[
D = {(x_i, y_i)}, \quad i = 1,2,...,n
]

### Step 1: Initialize prediction

Start with constant prediction:

[
\hat{y}_i^{(0)} = \text{mean}(y)
]

---

### Step 2: Add trees sequentially

For boosting round ( t = 1 \dots T ):

Add a new tree ( f_t(x) )

[
\hat{y}_i^{(t)} = \hat{y}_i^{(t-1)} + f_t(x_i)
]

Each tree tries to minimize the loss.

---

# 3️⃣ Mathematical Formulation

## Objective Function

XGBoost minimizes:

[
\mathcal{L} = \sum_{i=1}^{n} l(y_i, \hat{y}*i) + \sum*{t=1}^{T} \Omega(f_t)
]

Where:

* ( l ) = loss function (MSE, log-loss, etc.)
* ( \Omega(f_t) ) = regularization term

---

## Regularization Term

[
\Omega(f) = \gamma T + \frac{1}{2} \lambda \sum_{j=1}^{T} w_j^2
]

Where:

* ( T ) = number of leaves
* ( w_j ) = leaf weights
* ( \gamma ) = complexity penalty
* ( \lambda ) = L2 regularization

👉 This prevents overfitting.

---

# 4️⃣ Second-Order Taylor Approximation (Key Feature 🚀)

Unlike traditional gradient boosting, XGBoost uses:

* First derivative (gradient)
* Second derivative (Hessian)

Expand loss using Taylor approximation:

[
l(y_i, \hat{y}_i^{(t)}) \approx l(y_i, \hat{y}_i^{(t-1)}) + g_i f_t(x_i) + \frac{1}{2} h_i f_t^2(x_i)
]

Where:

[
g_i = \partial_{\hat{y}} l(y_i, \hat{y})
]

[
h_i = \partial^2_{\hat{y}} l(y_i, \hat{y})
]

This makes optimization faster and more accurate.

---

# 5️⃣ Tree Structure in XGBoost

Each tree maps input to leaf score:

[
f(x) = w_{q(x)}
]

Where:

* ( q(x) ) assigns sample to a leaf
* ( w ) is leaf weight

---

## Optimal Leaf Weight

For a leaf ( j ):

[
w_j^* = - \frac{G_j}{H_j + \lambda}
]

Where:

* ( G_j = \sum g_i )
* ( H_j = \sum h_i )

---

## Split Gain Formula (Very Important)

When splitting a node:

[
Gain = \frac{1}{2} \left(
\frac{G_L^2}{H_L + \lambda} +
\frac{G_R^2}{H_R + \lambda} -
\frac{G^2}{H + \lambda}
\right) - \gamma
]

If Gain > 0 → split accepted.

This is why XGBoost is very powerful.

---

# 6️⃣ Regularization Techniques in XGBoost

XGBoost includes:

* L1 Regularization (α)
* L2 Regularization (λ)
* Tree pruning
* Shrinkage (learning rate η)
* Column subsampling
* Row subsampling

This makes it more robust than traditional Gradient Boosting.

# 7️⃣ Important Hyperparameters

| ----------------- | --------------------------------- |
| n_estimators      | Number of trees                   |
| max_depth         | Maximum tree depth                |
| learning_rate (η) | Shrinks contribution of each tree |
| gamma             | Minimum split gain                |
| lambda            | L2 regularization                 |

| subsample         | Row sampling                      |

# 8️⃣ Why XGBoost is Powerful

✅ Uses second-order optimization
✅ Built-in regularization
✅ Handles missing values
✅ Parallel processing
✅ Tree pruning
✅ Highly optimized implementation

---

# 9️⃣ Classification vs Regression

### Regression:

Loss = Mean Squared Error

### Classification:

Loss = Logistic loss

Binary prediction:

[
P(y=1|x) = \sigma(\hat{y})
]

Where:

[
\sigma(z) = \frac{1}{1+e^{-z}}
]

---

# 🔟 Computational Complexity

For:

* n = samples
* d = features
* K = trees

Approx complexity:

[
O(K \cdot n \cdot d \log n)
]

Optimized using histogram-based splitting.
