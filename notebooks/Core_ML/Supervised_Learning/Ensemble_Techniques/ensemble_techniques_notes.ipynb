# Ensemble Techniques



---

## What are Ensemble Techniques?

**Ensemble learning** combines multiple models (often called *base learners* or *weak learners*) to produce a single, stronger model. The intuition is that different models make different mistakes, and combining them can cancel out individual errors.

Two of the most important ensemble families are **Bagging** and **Boosting**.

---

## 1. Bagging (Bootstrap Aggregating)

### Core idea

Train **many independent models in parallel**, each on a slightly different version of the dataset, and then **aggregate their predictions**.

### How it works

1. Create multiple datasets by **sampling with replacement** from the original training data (this is called *bootstrap sampling*).
2. Train the same model type on each dataset.
3. Combine predictions:

   * **Classification** → majority voting
   * **Regression** → averaging

### What problem it solves

Bagging mainly reduces **variance**.

* High-variance models (like decision trees) are very sensitive to small changes in data.
* Bagging stabilizes them by averaging out the noise.

### Key characteristics

* Models are trained **independently**
* Easy to **parallelize**
* Doesn’t focus on hard examples specifically

### Classic example: Random Forest

Random Forest is basically:

* Bagging + decision trees
* Extra randomness by selecting a **subset of features** at each split

### When Bagging works best

* Model has **high variance, low bias**
* Data is noisy
* Overfitting is a concern

---

## 2. Boosting

### Core idea

Train models **sequentially**, where each new model focuses more on the **mistakes made by previous models**.

### How it works (high level)

1. Train a first weak learner on the data.
2. Increase the importance (weight) of misclassified or high-error samples.
3. Train the next learner focusing more on those hard cases.
4. Combine all learners into a weighted sum or vote.

Each model is trying to “fix” what the previous ones got wrong.

### What problem it solves

Boosting mainly reduces **bias** (and sometimes variance too).

* It turns many **weak learners** into a strong one.

### Key characteristics

* Models are trained **sequentially**
* Harder to parallelize
* Sensitive to noisy data and outliers

---

## Popular Boosting Algorithms

### AdaBoost (Adaptive Boosting)

* Increases weights of misclassified samples
* Final prediction is a **weighted vote** of all learners
* Very sensitive to outliers

### Gradient Boosting

* Models are trained to predict the **residual errors** of the previous ensemble
* Uses gradient descent to minimize a loss function
* Very flexible and powerful

### XGBoost / LightGBM / CatBoost

* Optimized, scalable versions of gradient boosting
* Handle regularization, missing values, and large datasets efficiently
* Widely used in competitions and real-world systems

---

## Bagging vs Boosting (Side-by-Side)

| Aspect               | Bagging              | Boosting           |
| -------------------- | -------------------- | ------------------ |
| Training style       | Parallel             | Sequential         |
| Focus                | Reducing variance    | Reducing bias      |
| Data sampling        | Bootstrap sampling   | Reweighted samples |
| Sensitivity to noise | Low                  | High               |
| Typical base model   | High-variance models | Weak learners      |
| Example              | Random Forest        | AdaBoost, XGBoost  |

---

## Intuition in One Line

* **Bagging**: “Let’s train lots of models independently and average them.”
* **Boosting**: “Let’s keep training models that focus on what we keep getting wrong.”
