# Cross Validation

Cross-validation (CV) is a way to estimate how well your model will perform on unseen data.

Instead of training once and testing once, you:

1. Split the data into parts
2. Train on some parts
3. Test on the remaining part
4. Repeat this process multiple times
5. Average the results

This gives a more reliable performance estimate and helps with model selection and hyperparameter tuning (like choosing

1. **Hold-Out Validation (Train/Test Split)**
How it works:

Split data once (e.g., 80% train, 20% test)

Train on the training set

Evaluate on the test set
Pros:
- Simple
- Fast

Cons:
- High variance (depends heavily on the split)
- Not reliable for small datasets

2. **k-Fold Cross-Validation (Most Common)**
How it works:

Split data into k equal folds

For each fold:

- Train on k−1 folds
- Test on the remaining fold

Average the k results

Typical values: k = 5 or 10
Pros:
- Balanced bias–variance tradeoff
- Efficient use of data

Cons:
- More computation than hold-out


3. **Stratified k-Fold Cross-Validation**
How it works:

Same as k-fold, but each fold preserves the class proportions

Important for classification, especially with imbalanced data

Pros:
- More reliable class representation
- Less biased evaluation

Standard for classification problems.

4. **Leave-One-Out Cross-Validation (LOOCV)**
How it works:

Each data point is used once as the test set

Train on all remaining points

Repeat for every observation

Pros:
- Uses almost all data for training
- Low bias

Cons:
- Very expensive computationally
- High variance

Useful for very small datasets.

5. **Repeated k-Fold Cross-Validation**
How it works:

Perform k-fold CV multiple times with different random splits

Average all results

Pros:
- More stable estimates
- Reduces randomness

Cons:
- More computation

When you want extra reliability.

6. **Time Series Cross-Validation (Rolling / Forward Chaining)**
How it works:

Respect time order

Train on past data, test on future data

eg:

Train: [1 2 3] → Test: [4]

Train: [1 2 3 4] → Test: [5]


Pros:
- Realistic for time-dependent data

Cons:
- Fewer training samples early on

When you want extra reliability.

7. **Group k-Fold Cross-Validation**
How it works:

Data points belong to groups

All samples from a group appear in only one fold

Example:

Multiple measurements from the same person

Multiple images of the same object


Pros:
- Prevents data leakage

Essential when observations are not independent.

| Type              | Best For                    |
| ----------------- | --------------------------- |
| Hold-Out          | Large datasets              |
| k-Fold            | General purpose             |
| Stratified k-Fold | Classification (imbalanced) |
| LOOCV             | Very small datasets         |
| Repeated k-Fold   | Stable evaluation           |
| Time Series CV    | Temporal data               |
| Group k-Fold      | Grouped/dependent data      |
