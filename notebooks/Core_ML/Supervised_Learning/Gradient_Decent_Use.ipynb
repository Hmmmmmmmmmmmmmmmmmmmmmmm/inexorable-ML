### **1. Logistic Regression**

**GD in Logistic Regression**:

* **Purpose**: In logistic regression, Gradient Descent is used to find the **optimal parameters** (weights) for the model. The goal is to minimize the **log loss** (binary cross-entropy) function.

* **Mathematics**:
  [
  J(\theta) = - \frac{1}{m} \sum_{i=1}^{m} \left[ y_i \log(h_\theta(x_i)) + (1 - y_i) \log(1 - h_\theta(x_i)) \right]
  ]
  where ( h_\theta(x) = \frac{1}{1 + e^{-\theta^T x}} ) is the sigmoid function.

* **Process**:

  * Start with random weights ( \theta ).
  * Compute the gradient of the loss function with respect to ( \theta ) using:
    [
    \nabla_\theta J(\theta) = \frac{1}{m} \sum_{i=1}^{m} (h_\theta(x_i) - y_i) x_i
    ]
  * Update the weights using the formula:
    [
    \theta := \theta - \alpha \nabla_\theta J(\theta)
    ]
    where ( \alpha ) is the learning rate.

* **Why Gradient Descent**: Logistic regression is a linear model, but the output is transformed via the sigmoid function, making the optimization process non-linear. GD helps minimize the log loss to find the optimal weights for the model.

### **2. Support Vector Machines (SVM)**

**GD in SVM**:

* **Purpose**: In SVM, Gradient Descent can be used to **optimize the objective function**, which is the margin between the classes while also minimizing classification errors (hinge loss).

* **Mathematics**:
  The objective is to maximize the margin, which is the inverse of the norm of the weight vector ( |\mathbf{w}| ). The loss function includes a term for the margin and a regularization term for classification errors (hinge loss):
  [
  L(w, b) = \frac{1}{2} |w|^2 + C \sum_{i=1}^{n} \max(0, 1 - y_i (w^T x_i + b))
  ]
  where ( C ) is a regularization parameter.

* **Process**:

  * Compute the gradient of the loss function with respect to ( w ) and ( b ).
  * Update the parameters iteratively using:
    [
    w := w - \alpha \frac{\partial L}{\partial w}, \quad b := b - \alpha \frac{\partial L}{\partial b}
    ]
  * Repeat until convergence.

* **Why Gradient Descent**: The hinge loss function (used in SVMs) is not differentiable at some points, but GD still works in practice by using subgradients. GD allows us to find the optimal decision boundary that maximizes the margin.

### **3. Naive Bayes Theorem**

**GD in Naive Bayes**:

* **Purpose**: Naive Bayes is based on probabilistic reasoning and typically doesn't use Gradient Descent for parameter estimation (since it's based on closed-form solutions for parameter estimation like **Maximum Likelihood Estimation**).

* **Mathematics**:
  The parameters in Naive Bayes are estimated by maximizing the likelihood of the observed data:
  [
  P(y|x) = \frac{P(y) \prod_{i=1}^{n} P(x_i | y)}{P(x)}
  ]
  where ( P(y|x) ) is the posterior probability.

* **Why Gradient Descent**: Naive Bayes typically avoids gradient descent and uses simpler probabilistic reasoning for parameter estimation. However, if you extend Naive Bayes to more complex variants (like **Gaussian Naive Bayes**), gradient-based optimization could be used for learning the parameters if necessary.

---

### **4. K-Nearest Neighbors (KNN)**

**GD in KNN**:

* **Purpose**: KNN is a **non-parametric** algorithm, meaning it doesn't learn a model in the traditional sense. There is no need for Gradient Descent in KNN because the algorithm makes predictions based on the closest training examples in the feature space.

* **Why No Gradient Descent**: Since KNN doesn’t involve fitting a model or estimating parameters (it simply stores the training data and makes predictions by finding the nearest neighbors), gradient descent isn’t used in this algorithm.

---

### **5. Decision Tree Classifier and Regressor**

**GD in Decision Trees**:

* **Purpose**: Decision trees build a model by **splitting** the data at each node based on the feature that reduces the most **impurity** (e.g., Gini index or entropy for classification, or MSE for regression).

* **Mathematics**:
  For classification:
  [
  Gini(t) = 1 - \sum_{k=1}^{K} p_k^2
  ]
  where ( p_k ) is the proportion of class ( k ) in a node ( t ).

* **Why No Gradient Descent**: Decision trees typically **do not use gradient descent**. Instead, they use a **greedy algorithm** to select the best split at each node. At each split, the algorithm evaluates potential features and chooses the one that results in the most significant reduction in impurity. This process continues until stopping conditions are met.

---

### **6. Random Forest Machine Learning**

**GD in Random Forest**:

* **Purpose**: Like decision trees, Random Forests are an ensemble method of decision trees. Each tree is built independently (using bootstrapped data), and then the final prediction is made by aggregating the predictions of all trees.

* **Why No Gradient Descent**: Random Forests also use **decision trees** as base learners, which don't use gradient descent. Instead, they use a **bagging** approach, where each tree is grown independently with bootstrapped data and the final prediction is an average (regression) or majority vote (classification).

---

### **7. AdaBoost Machine Learning Algorithm**

**GD in AdaBoost**:

* **Purpose**: AdaBoost combines several weak learners (often decision trees) in a **sequential manner** to form a strong model. Each subsequent learner tries to correct the mistakes made by the previous one by assigning higher weights to misclassified examples.

* **Mathematics**:
  The algorithm assigns weights to each training example and updates these weights at each iteration based on the errors made by the weak learner. The final prediction is a weighted vote of all the weak learners.

* **Why No Gradient Descent**: AdaBoost doesn’t directly use gradient descent. It uses a **boosting** technique, where each weak learner is trained to minimize the errors of the ensemble’s previous learners, but the optimization process is based on the re-weighting of data points, not gradient descent.

---

### **8. Gradient Boosting Machine (GBM)**

**GD in Gradient Boosting**:

* **Purpose**: Gradient Boosting optimizes the model by fitting new weak learners (usually decision trees) on the residuals (errors) of the previous learners. This process is guided by **gradient descent**.

* **Mathematics**:

  * The loss function ( L(y, F(x)) ) (e.g., MSE or log loss) is minimized.
  * In each iteration, a new tree is trained on the **negative gradient** of the loss function:
    [
    \nabla_\theta L(y, F(x)) = - \frac{\partial}{\partial F(x)} L(y, F(x))
    ]
  * The trees are fitted to the gradients (residuals) from the previous iterations, and the predictions are updated iteratively:
    [
    F_m(x) = F_{m-1}(x) + \eta h_m(x)
    ]

* **Why Gradient Descent**: Gradient Boosting uses **gradient descent** to iteratively minimize the loss function by updating the model in the direction that reduces the residuals (errors). The key difference from AdaBoost is that the new trees are directly optimized based on the gradient of the loss function, allowing more efficient learning.

---

### **Summary**:

* **Gradient Descent** is used in algorithms like **Logistic Regression, SVMs, and Gradient Boosting** to minimize a **loss function** iteratively by updating model parameters.
* **Non-parametric algorithms** like **KNN** and **Naive Bayes** don’t require gradient descent since they don’t involve optimizing model parameters.
* **Ensemble methods** like **Random Forests** and **AdaBoost** also don’t use gradient descent directly but instead employ other techniques like **bagging** and **boosting** for optimization.

**Gradient Descent** is key in optimization, especially when the model involves continuous parameters that need to be optimized over a dataset,


such as in **Logistic Regression**, **SVM**, and **Gradient Boosting**.
