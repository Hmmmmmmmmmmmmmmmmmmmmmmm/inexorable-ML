# Simple Linear Regression

## 1. Simple Linear Regression (SLR)
**Core Intuition:** SLR models the relationship between a single independent feature and a dependent target by fitting a straight line that minimizes the error between observed and predicted values.

**The Hypothesis Equation:**
$$h_\theta(x) = \beta_0 + \beta_1 x$$

* **$y$ (Target):** The actual value we observe.
* **$x$ (Feature):** The input variable.
* **$\beta_0$ (Intercept/Bias):** The baseline value of $y$ when $x=0$.
* **$\beta_1$ (Slope/Weight):** The "impact" of $x$. If $\beta_1 = 3$, a 1-unit increase in $x$ leads to a 3-unit increase in $y$.
* **$\epsilon$ (Error Term):** The residual difference (noise) that the model cannot explain.

---

## 2. Cost Function & Optimization
To determine which line is "best," we need a mathematical way to quantify error.

**The Cost Function ($J$):**
We typically use **Mean Squared Error (MSE)**. We square the errors to (1) penalize outliers heavily and (2) ensure the error surface is convex (bowl-shaped), ensuring a single global minimum.

$$J(\beta_0, \beta_1) = \frac{1}{2m} \sum_{i=1}^{m} (h_\theta(x^{(i)}) - y^{(i)})^2$$

*(Note: The factor $\frac{1}{2}$ is a mathematical convenience to cancel out the exponent when taking the derivative).*

**Convergence Algorithm (Gradient Descent):**
Gradient Descent is the iterative engine that minimizes the Cost Function.
1.  **Initialize:** Start with random $\beta$ values (often 0).
2.  **Calculate Slope:** Find the gradient (derivative) of the cost curve at the current point.
3.  **Update:** Step in the opposite direction of the gradient to roll down the "bowl."

**Update Rule:**
$$\beta_{new} = \beta_{old} - \alpha \times \text{gradient}$$
* **$\alpha$ (Learning Rate):** Defines the step size.
    * *Small $\alpha$:* Slow convergence.
    * *Large $\alpha$:* Risk of overshooting the minimum.

---

## 3. Multiple Linear Regression & OLS
When using multiple features ($x_1, x_2, ..., x_n$), the line becomes a **hyperplane**.

**The Equation:**
$$y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + ... + \beta_n x_n$$

**Ordinary Least Squares (OLS):**
Unlike the iterative Gradient Descent, OLS is an analytical, "closed-form" solution that uses Linear Algebra to find the perfect coefficients instantly.

$$\beta = (X^T X)^{-1} X^T y$$

* **Assumption:** Columns in $X$ must be independent (no multi-collinearity).
* **Limitation:** Computationally expensive ($O(n^3)$) for datasets with huge numbers of features due to matrix inversion.

---

## 4. Performance Metrics
Accuracy is not used in regression; we quantify "distance" from the truth.

| Metric | Formula | Professional Usage |
| :--- | :--- | :--- |
| **MAE** (Mean Absolute Error) | $\frac{1}{n}\sum |y_i - \hat{y}_i|$ | Robust to outliers. Easy to interpret (avg error in original units). |
| **MSE** (Mean Squared Error) | $\frac{1}{n}\sum (y_i - \hat{y}_i)^2$ | Heavily penalizes large errors. Differentiable (good for optimization). |
| **RMSE** (Root MSE) | $\sqrt{MSE}$ | The standard industry metric. Interpretable units like MAE, but penalizes large errors like MSE. |
| **$R^2$ Score** | $1 - \frac{SS_{res}}{SS_{tot}}$ | "Goodness of fit." 0 to 1. Explains what percentage of variance comes from the model vs. the average mean. |
| **Adjusted $R^2$** | $1 - (1-R^2)\frac{n-1}{n-p-1}$ | **Critical for Multiple Regression.** Penalizes the score if you add useless features that don't improve the model. |

---

## 5. Overfitting vs. Underfitting (Bias-Variance Tradeoff)
* **Underfitting (High Bias):** Model is too simple. It cannot capture the pattern (e.g., fitting a line to a curve).
    * *Result:* Poor performance on both Training and Testing data.
* **Overfitting (High Variance):** Model is too complex. It memorizes the noise in the training data.
    * *Result:* Great Training score, terrible Testing score.
* **Goal:** A **Generalized** model that performs well on unseen data.

---

## 6. Polynomial Regression
Used when data shows a non-linear pattern (e.g., a curve).

**Intuition:**
We do **not** change the regression algorithm to be non-linear. Instead, we transform the **input features** to be non-linear (e.g., $x^2, x^3$) and feed them into a standard linear model.

$$y = \beta_0 + \beta_1 x + \beta_2 x^2 + ... + \beta_n x^n$$

**Practical Implementation (The Pipeline):**
In production code (Scikit-Learn), this is a 3-step process:
1.  **PolynomialFeatures:** Generates higher-order terms ($x \rightarrow x, x^2$).
2.  **StandardScaler:** Normalizes data. *Crucial* because $x^2$ grows much faster than $x$, which can break Gradient Descent.
3.  **LinearRegression:** Fits the model on the transformed, scaled data.

**The Risk:**
Increasing the polynomial degree too high (e.g., degree=20) guarantees **Overfitting**, as the curve will oscillate wildly to pass through every noise point.

```python
# dividing independent and dependent feature
x = df[['Weight']] # independent feature must be dataframe or 2D array
type(x)
```

```python
y=df['Height'] # output can be 1D or Series
type(y)
```

```python
y
```

```python
# train test split
from sklearn.model_selection import train_test_split
```

```python
#keeping 75:25 ratio
X_train, X_test, y_train, y_test = train_test_split(x,y,test_size = 0.25,random_state=42)
```

```python
X_train.shape
```

```python
#Standardization
#since height and weights are of different units
#and since we are doing regression GD will take time
#thus we will take z_score
from sklearn.preprocessing import StandardScaler
```

- fit = learn the rules
- transform = apply the rules
- Test data must follow training rules, not invent new ones

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train) #we dont want data leakage so we use
# the same fit and transform the test data else the test data will
#have an unfair advantage for the training as it is not possible
#to calculate the mean, SD (fit calculation) every time, thus,
X_test = scaler.transform(X_test)
```

```python
#now applying simple linear regression
from sklearn.linear_model import LinearRegression
```

```python
regression = LinearRegression()
```

```python
regression.fit(X_train,y_train)
```

`fit_intercept=True`
- Model learns an intercept (**b**)
- Equation: `y = mx + b`
- Almost always keep this **True**

`copy_X=True`
- Makes a copy of `X_train` internally
- Prevents accidental modification of your data
- You almost never change this

`tol=1e-06`
- Numerical tolerance for stopping the solver
- Smaller value = more precise
- You can ignore this for now

`n_jobs=None`
- Number of CPU cores to use
- `None` = default behavior (usually 1 core)

`positive=False`
- If `True`, forces the slope to be positive
- Normally leave this **False**

```python
regression.coef_
```

```python
regression.get_params()
```

```python
print("Slope (m):", regression.coef_)
print("Intercept (b):", regression.intercept_)
```

here we see that 1 unit in weight value leads to slope change in height
(coefficien tdef)

| Theory Step                                   | sklearn Behind the Scenes                                    |
| --------------------------------------------- | ------------------------------------------------------------ |
| Calculate slope and intercept to minimize RSS | Uses linear algebra to solve (\hat{\beta} = (X^TX)^{-1}X^Ty) |
| Check if intercept is needed                  | `fit_intercept=True` triggers adding a column of 1s to X     |
| Store coefficients                            | Saves in `.coef_` and `.intercept_` attributes               |
| Prepare model for predictions                 | `.predict()` will use (y = X\hat{\beta} + \beta_0)           |


```python
#plotting best-fit line
plt.scatter(df['Weight'],df['Height'])
plt.plot(X_train,regression.predict(X_train),color='red')
```

```python
#plotting best-fit line
plt.scatter(X_train,y_train)
plt.plot(X_train,regression.predict(X_train),color='red')
```

| Step                    | What happens                                 | Purpose                          |
| ----------------------- | -------------------------------------------- | -------------------------------- |
| `fit(X_train, y_train)` | Model learns coefficients from training data | Learn relationship between x & y |
| `predict(X_train)`      | Model predicts on **training data**          | Evaluate fit on training set     |
| `predict(X_test)`       | Model predicts on **new/unseen test data**   | Evaluate generalization          |

### R square
Formula

**R^2 = 1 - SSR/SST**

R^2 = coefficient of determination SSR = sum of squares of residuals SST = total sum of squares

```python
from sklearn.metrics import r2_score
```

```python
score = r2_score(y_test,y_pred)
score
```

**Adjusted R2 = 1 -[(1-R2)*(n-1)/(n-k-1)]**

where:
R2: The R2 of the model n: The number of observations k: The number of predictor variables

```python
1-(1-score)*(len(y_test)-1)/(len(y_test)-X_test.shape[1]-1)
```
