# Naive Bayes' Algo


---

## What is Naive Bayes?

**Naive Bayes** is a **probabilistic machine learning algorithm** based on **Bayes’ Theorem**. It’s mainly used for **classification tasks**, especially when working with **text data**.

Despite being called *“naive”*, it’s surprisingly powerful and fast.

---

## The Core Idea (Bayes’ Theorem)

At its heart, Naive Bayes answers:

> **“Given some features, what’s the probability this data belongs to a certain class?”**

Bayes’ Theorem:

[
P(Class \mid Features) = \frac{P(Features \mid Class) \cdot P(Class)}{P(Features)}
]

Naive Bayes uses this to pick the class with the **highest probability**.

---

## Why is it called “Naive”?

Because it makes a **big assumption**:

> **All features are independent of each other given the class**

Example (clearly unrealistic but works):

* In spam detection, it assumes the words *“free”* and *“money”* are independent.
* In reality, they’re related—but Naive Bayes ignores that.

Surprisingly, this assumption works well in practice 🤯

---

## How the Algorithm Works (Step-by-Step)

1. **Calculate prior probability** of each class

   * e.g., P(Spam), P(Not Spam)

2. **Calculate likelihood**

   * Probability of each feature given a class
   * e.g., P("free" | Spam)

3. **Apply Bayes’ formula** for each class

4. **Choose the class with the highest probability**

---

## Types of Naive Bayes

Different versions depending on data type:

### 1. **Gaussian Naive Bayes**

* Used for **continuous numerical data**
* Assumes features follow a **normal distribution**
* Common in medical & sensor data

### 2. **Multinomial Naive Bayes**

* Used for **text classification**
* Works with word counts or frequencies
* Very popular for **spam detection**, **sentiment analysis**

### 3. **Bernoulli Naive Bayes**

* Uses **binary features** (0 or 1)
* Example: word present or not present

---

## Where Naive Bayes Shines

* Text classification (emails, reviews, tweets)
* Spam filtering
* Sentiment analysis
* Document categorization
* Real-time prediction (super fast)

---

## Pros

* Simple and easy to implement
* Extremely fast (great for large datasets)
* Works well with high-dimensional data
* Needs relatively little training data

---

## Cons

* Independence assumption is often unrealistic
* Can perform poorly when features are strongly correlated
* Zero probability problem (fixed with **Laplace smoothing**)

---

## A Tiny Example

Imagine classifying an email:

> “Win free money”

Naive Bayes calculates:

* Probability it’s spam
* Probability it’s not spam
  …and picks whichever is higher.

---

## TL;DR

* Naive Bayes = **probability-based classifier**
* Assumes feature independence
* Fast, simple, and great for text
* “Naive” but surprisingly effective

```python
X,y = load_iris(
    return_X_y=True
)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X,y,
    random_state=67,
    test_size=0.30
)
```

```python
X_train
```

```python
from sklearn.naive_bayes import GaussianNB
```

```python
gnb = GaussianNB()
```

```python
gnb.fit(X_train, y_train)
```

```python
y_pred = gnb.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
print(confusion_matrix(y_true=y_test, y_pred=y_pred))
print(accuracy_score(y_true=y_test, y_pred=y_pred))
print(classification_report(y_true=y_test, y_pred=y_pred))
```

```python
from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(
    gnb,
    X,
    y,
    cv=7,
    scoring='accuracy'
)

print("CV Accuracy Scores:", cv_scores)
print("Mean CV Accuracy:", cv_scores.mean())
print("Std Dev:", cv_scores.std())
```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score
from sklearn.naive_bayes import GaussianNB
import numpy as np

skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=3)
gnb = GaussianNB()

accuracies = []

for train_index, test_index in skf.split(X, y):
    X_train, X_test = X[train_index], X[test_index]
    y_train, y_test = y[train_index], y[test_index]

    gnb.fit(X_train, y_train)
    y_pred = gnb.predict(X_test)

    accuracies.append(accuracy_score(y_test, y_pred))

print("Fold Accuracies:", accuracies)
print("Mean Accuracy:", np.mean(accuracies))
```
