*Now we try HyperParameter Tunning using optimal para_grid*

✅ C (regularization)

Log-scale catches under/over-regularization

1000 helps when data is very clean

✅ gamma (rbf/poly)

1e-3 → 1e-1 covers smooth → complex boundaries

"scale" still often wins

✅ coef0 (poly)

Controls influence of higher-order terms

Makes a huge difference for poly kernels

✅ class_weight

Saves you if the dataset is imbalanced

Often boosts recall dramatically

```python
from sklearn.model_selection import GridSearchCV
```

```python
svm_mew = SVC()
```

```python
param_grid = [
    # Linear kernel
    {
        "kernel": ["linear"],
        "C": [0.01, 0.1, 1, 10, 100, 1000],
        "class_weight": [None, "balanced"]
    },

    # RBF kernel (most powerful)
    {
        "kernel": ["rbf"],
        "C": [0.1, 1, 10, 100, 1000],
        "gamma": ["scale", "auto", 1e-3, 1e-2, 1e-1],
        "class_weight": [None, "balanced"]
    },

    # Polynomial kernel
    {
        "kernel": ["poly"],
        "C": [0.1, 1, 10, 100],
        "gamma": ["scale", "auto"],
        "degree": [2, 3, 4, 5],
        "coef0": [0, 0.5, 1],
        "class_weight": [None, "balanced"]
    }
]
```

```python
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold()
#GridSearchCV
from sklearn.model_selection import GridSearchCV
grid = GridSearchCV(estimator= svm_mew,
                    param_grid= param_grid,
                    scoring='accuracy',
                    cv=cv,
                    n_jobs=-1,
                    verbose=1
                    )
```

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.svm import SVC
from scipy.stats import loguniform, randint, uniform

# Define the parameter distributions
param_dist = [
    # Linear kernel
    {
        "kernel": ["linear"],
        "C": loguniform(1e-2, 1e3),   # explore small to large regularization
        # "class_weight": [None, "balanced"]
    },

    # RBF kernel
    {
        "kernel": ["rbf"],
        "C": loguniform(1e-1, 1e3),
        "gamma": loguniform(1e-4, 1e-1),
        # "class_weight": [None, "balanced"]
    },

    # Polynomial kernel
    {
        "kernel": ["poly"],
        "C": loguniform(1e-1, 1e2),
        "gamma": loguniform(1e-4, 1e-1),
        "degree": randint(2, 6),      # 2,3,4,5
        "coef0": uniform(0, 1),       # 0 → 1
        # "class_weight": [None, "balanced"]
    }
]
```

```python
# SVC model
svc = SVC()

# Randomized search
random_search = RandomizedSearchCV(
    estimator=svc,
    param_distributions=param_dist,
    n_iter=30,        # number of random combinations to try
    cv=5,
    scoring="accuracy",
    random_state=42,
    n_jobs=-1
)

# Fit to your training data
random_search.fit(X_train, y_train)
```

```python
# Best params and score
print(random_search.best_params_)
print(random_search.best_score_)
```

```python
y_pred_random_search = random_search.predict(X_test)
score = accuracy_score(y_pred_random_search, y_test)
print(score)
print(classification_report(y_pred_random_search, y_test))
print(confusion_matrix(y_pred_random_search,y_test))
```

*Trying a basic classification problem*

*-In titanic dataset i was getting roughly 0.88 acc given imbalance using Logistic_Regression, now trying the same cleaned dataset with SVC*

```python
df_t = pd.read_csv("../../../../data/titanic/cleaned_data_train_titanic.csv")
```

```python
df_t.head()
```

```python
X = df_t.drop("Survived", axis = 1)
y = df_t["Survived"]
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train, y_test = train_test_split(
    X,y,test_size = 0.25,
    random_state = 67,
    stratify=y
)
from sklearn.svm import SVC
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
svm_titanic = SVC(
            class_weight="balanced",
            random_state=42,
            probability=True
        )
```

```python
from scipy.stats import loguniform, randint, uniform

param_dist = [
    # Linear kernel (baseline)
    {
        "kernel": ["linear"],
        "C": loguniform(1e-2, 1e2),
    },

    # RBF kernel (most important)
    {
        "kernel": ["rbf"],
        "C": loguniform(1e-2, 1e3),
        "gamma": loguniform(1e-4, 1e-1),
    },

    # Polynomial kernel (controlled, otherwise explodes)
    {
        "kernel": ["poly"],
        "C": loguniform(1e-2, 1e2),
        "gamma": loguniform(1e-4, 1e-2),
        "degree": randint(2, 5),   # 2–4 only (5+ is rarely useful)
        "coef0": uniform(0, 1),
    }
]
```

```python
search = RandomizedSearchCV(
    svm_titanic,
    param_distributions=param_dist,
    n_iter=40,
    scoring="roc_auc",        # or "average_precision"
    cv=5,
    n_jobs=-1,
    verbose=1,
    random_state=42,
)
```

```python
search.fit(X_train_scaled, y_train)
```

```python
# Best params and score
print(search.best_params_)
print(search.best_score_)
y_pred_random_search = search.predict(X_test_scaled)
score = accuracy_score(y_pred_random_search, y_test)
print(score)
print(classification_report(y_test,y_pred_random_search))
print(confusion_matrix(y_test,y_pred_random_search))
```

```python
from sklearn.metrics import roc_curve, auc

y_probs = search.best_estimator_.predict_proba(X_test_scaled)[:, 1]

fpr, tpr, thresholds = roc_curve(y_test, y_probs)
roc_auc = auc(fpr, tpr)

print("ROC AUC:", roc_auc)
```

```python
plt.figure(figsize=(6, 5))
plt.plot(fpr, tpr, label=f"SVC (AUC = {roc_auc:.3f})")
plt.plot([0, 1], [0, 1], "k--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()
```

```python
from sklearn.metrics import precision_recall_curve, average_precision_score

precision, recall, thresholds = precision_recall_curve(y_test, y_probs)
ap = average_precision_score(y_test, y_probs)

print("PR AUC:", ap)
```

```python
plt.figure(figsize=(6, 5))
plt.plot(recall, precision, label=f"PR AUC = {ap:.3f}")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision–Recall Curve")
plt.legend()
plt.show()
```

```python
from sklearn.metrics import f1_score

f1_scores = [f1_score(y_test, y_probs >= t) for t in thresholds]
best_threshold = thresholds[np.argmax(f1_scores)]

print("Best threshold:", best_threshold)
```

```python
y_pred_opt = (y_probs >= best_threshold).astype(int)

print(classification_report(y_test, y_pred_opt))
print(confusion_matrix(y_test, y_pred_opt))
```
