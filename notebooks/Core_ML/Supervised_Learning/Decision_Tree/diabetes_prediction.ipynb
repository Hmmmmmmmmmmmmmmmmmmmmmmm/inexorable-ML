# Diabetes Prediction (DT)

```python
from sklearn.datasets import load_diabetes
```

```python
ds = load_diabetes()
ds
```

```python
ds.keys()
```

```python
print(ds['DESCR'])
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = ds['data']
data
```

```python
X = pd.DataFrame(data,columns=ds['feature_names'])
```

```python
X
```

```python
y = ds['target']
y
```

```python
X.describe()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20, random_state=67)
```

```python
sns.heatmap(X_train.corr(), annot=True)
```

```python
from sklearn.tree import DecisionTreeRegressor
dr = DecisionTreeRegressor()
```

```python
dr.fit(X_train,y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(dr,filled=True,rounded=True)
```

```python
y_pred=dr.predict(X_test)
from sklearn.metrics import r2_score,mean_absolute_error, mean_squared_error
print(r2_score(y_test,y_pred))
print(mean_absolute_error(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
```

```python
from sklearn.tree import DecisionTreeRegressor
dr = DecisionTreeRegressor()
from sklearn.model_selection import GridSearchCV
param_grid = {
    'criterion': ['squared_error', 'friedman_mse', 'absolute_error' ],
    'splitter': ['best', 'random' ],
    'max_depth': [1, 2, 3, 4, 5,],
}
grid = GridSearchCV(
    dr,
    param_grid=param_grid,
    cv=5,
    n_jobs=-1,
    verbose=2,
    scoring='r2'
    # scoring='neg_mean_squared_error'
)


grid.fit(X_train,y_train)
```

```python
best_tree = grid.best_estimator_

y_pred=best_tree.predict(X_test)
from sklearn.metrics import r2_score,mean_absolute_error, mean_squared_error
print(r2_score(y_test,y_pred))
print(mean_absolute_error(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
```

```python
plt.figure(figsize=(15,10))
tree.plot_tree(best_tree, filled=True, rounded=True)
plt.show()
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import GridSearchCV
from sklearn import tree
import matplotlib.pyplot as plt

param_grid = {
    'criterion': ['squared_error', 'friedman_mse', 'absolute_error'],
    'splitter': ['best', 'random'],
    'max_depth': [1, 2, 3, 4, 5]
}

dr = DecisionTreeRegressor()
grid = GridSearchCV(dr, param_grid, cv=5, n_jobs=-1, scoring='r2')
grid.fit(X_train, y_train)
print(grid.best_params_)
best_tree = grid.best_estimator_

y_pred = best_tree.predict(X_test)

print("R2:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
print("MSE:", mean_squared_error(y_test, y_pred))

plt.figure(figsize=(15,10))
tree.plot_tree(best_tree, filled=True, rounded=True)
plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=200,
    max_depth=5,
    random_state=42
)
rf.fit(X_train, y_train)

y_pred = rf.predict(X_test)

print("R2:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
print("MSE:", mean_squared_error(y_test, y_pred))

# plt.figure(figsize=(15,10))
# tree.plot_tree(best_tree, filled=True, rounded=True)
# plt.show()
```

Linear Regression achieved the highest R² score (0.537), outperforming tree-based and ensemble models. This suggests that the relationship between features and target is predominantly linear. | Metric | Question it answers               |
| ------ | --------------------------------- |
| R²     | How much structure did I explain? |
| MAE    | How wrong am I on average?        |
| MSE    | How bad are my worst mistakes?    |
