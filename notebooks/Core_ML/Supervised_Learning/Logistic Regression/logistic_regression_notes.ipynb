# Logistic Regression


## 1. Introduction
Logistic Regression is a **supervised learning algorithm** primarily used for **binary classification** problems. Despite its name, it is a **classification** algorithm, not a regression one. It models the probability that a given input belongs to a particular class.

---

## 2. Theoretical Definition
Logistic Regression estimates the probability that a dependent variable `y` belongs to a class (usually `1`) given independent features `X`.

Formally:
\[
P(y = 1 | X) = \sigma(z)
\]
where:
\[
z = w^T X + b
\]

- `w` → weight vector  
- `X` → feature vector  
- `b` → bias (intercept)  
- `σ(z)` → sigmoid (logistic) function  

---

## 3. Sigmoid (Logistic) Function
The sigmoid function maps any real-valued number into the range `(0, 1)`.

\[
\sigma(z) = \frac{1}{1 + e^{-z}}
\]

### Properties:
- Output close to `1` → high confidence for class `1`
- Output close to `0` → high confidence for class `0`
- S-shaped curve
- Differentiable (important for optimization)

---

## 4. Decision Boundary
Logistic Regression predicts:
\[
\hat{y} =
\begin{cases}
1 & \text{if } \sigma(z) \ge 0.5 \\
0 & \text{otherwise}
\end{cases}
\]

Since `σ(z) ≥ 0.5` when `z ≥ 0`, the **decision boundary** is:
\[
w^T X + b = 0
\]

This boundary is:
- Linear in feature space
- A line (2D), plane (3D), or hyperplane (higher dimensions)

---

## 5. Mathematical Working

### 5.1 Linear Combination
Each feature contributes linearly:
\[
z = w_1x_1 + w_2x_2 + \dots + w_nx_n + b
\]

- Positive weight → increases probability
- Negative weight → decreases probability
- Magnitude → strength of influence

---

### 5.2 Likelihood Function
For `m` training samples:

\[
L(w, b) = \prod_{i=1}^{m} P(y^{(i)} | x^{(i)})
\]

\[
= \prod_{i=1}^{m} (\hat{y}^{(i)})^{y^{(i)}} (1 - \hat{y}^{(i)})^{(1 - y^{(i)})}
\]

---

### 5.3 Log-Likelihood
Taking logarithm for numerical stability:

\[
\ell(w, b) = \sum_{i=1}^{m} \left[
y^{(i)} \log(\hat{y}^{(i)}) +
(1 - y^{(i)}) \log(1 - \hat{y}^{(i)})
\right]
\]

---

### 5.4 Cost Function (Binary Cross-Entropy)
We minimize the **negative log-likelihood**:

\[
J(w, b) = -\frac{1}{m} \sum_{i=1}^{m}
\left[
y^{(i)} \log(\hat{y}^{(i)}) +
(1 - y^{(i)}) \log(1 - \hat{y}^{(i)})
\right]
\]

This is also called **Log Loss**.

---

## 6. Optimization (Gradient Descent)

### 6.1 Gradients
Partial derivatives:

\[
\frac{\partial J}{\partial w} = \frac{1}{m} X^T (\hat{y} - y)
\]

\[
\frac{\partial J}{\partial b} = \frac{1}{m} \sum (\hat{y} - y)
\]

---

### 6.2 Parameter Update
\[
w := w - \alpha \frac{\partial J}{\partial w}
\]
\[
b := b - \alpha \frac{\partial J}{\partial b}
\]

- `α` → learning rate

---

## 7. Feature Handling in Logistic Regression

### 7.1 Feature Scaling
Highly recommended:
- Standardization (Z-score)
- Min-Max scaling

Reason:
- Faster convergence
- Stable gradients

---

### 7.2 Feature Importance
- Weight magnitude indicates importance
- Larger absolute value → stronger influence
- Sign indicates direction of influence

---

### 7.3 Categorical Features
Handled via:
- One-Hot Encoding
- Binary Encoding

---

### 7.4 Interaction Features
Logistic Regression is linear, but can handle non-linearity by:
- Polynomial features
- Feature interactions (e.g., `x1 * x2`)

---

## 8. Regularization
Used to prevent overfitting.

### 8.1 L2 Regularization (Ridge)
\[
J = J + \lambda \sum w^2
\]

- Shrinks weights
- Keeps all features

---

### 8.2 L1 Regularization (Lasso)
\[
J = J + \lambda \sum |w|
\]

- Encourages sparsity
- Performs feature selection

---

### 8.3 Elastic Net
Combination of L1 and L2.

---

## 9. Implementation (From Scratch)

```python
import numpy as np

class LogisticRegression:
    def __init__(self, lr=0.01, epochs=1000):
        self.lr = lr
        self.epochs = epochs

    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    def fit(self, X, y):
        m, n = X.shape
        self.w = np.zeros(n)
        self.b = 0

        for _ in range(self.epochs):
            z = np.dot(X, self.w) + self.b
            y_hat = self.sigmoid(z)

            dw = (1/m) * np.dot(X.T, (y_hat - y))
            db = (1/m) * np.sum(y_hat - y)

            self.w -= self.lr * dw
            self.b -= self.lr * db

    def predict(self, X):
        z = np.dot(X, self.w) + self.b
        y_hat = self.sigmoid(z)
        return (y_hat >= 0.5).astype(int)


## 10. Implementation Using Scikit-Learn

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(penalty='l2', solver='lbfgs')
model.fit(X_train, y_train)

y_pred = model.predict(X_test)


11. Assumptions

Linear relationship between features and log-odds

Independent observations

Low multicollinearity

Large sample size preferred

12. Advantages

Simple and interpretable

Probabilistic output

Fast training

Works well for linearly separable data

13. Limitations

Cannot model complex non-linear relationships directly

Sensitive to outliers

Performance drops with highly correlated features

14. Multiclass Logistic Regression

Handled using:

One-vs-Rest (OvR)

Softmax Regression

Softmax:

𝑃
(
𝑦
=
𝑘
∣
𝑥
)
=
𝑒
𝑤
𝑘
𝑇
𝑥
∑
𝑗
𝑒
𝑤
𝑗
𝑇
𝑥
P(y=k∣x)=
∑
j
	​

e
w
j
T
	​

x
e
w
k
T
	​

x
	​

15. Summary

Logistic Regression models log-odds linearly

Uses sigmoid to map outputs to probabilities

Optimized using cross-entropy loss

Feature handling and regularization are crucial for performance

# Performance Metrics for Classification — Detailed Notes

## 1. Introduction
Performance metrics for classification are used to **evaluate how well a classification model performs**. Different metrics highlight different aspects of model behavior, especially in cases of **class imbalance**.

---

## 2. Confusion Matrix
The confusion matrix summarizes prediction results.

| Actual \ Predicted | Positive (1) | Negative (0) |
|--------------------|--------------|--------------|
| Positive (1)       | TP           | FN           |
| Negative (0)       | FP           | TN           |

### Definitions:
- **TP (True Positive):** Correctly predicted positive
- **TN (True Negative):** Correctly predicted negative
- **FP (False Positive):** Incorrectly predicted positive
- **FN (False Negative):** Incorrectly predicted negative

---

## 3. Accuracy
Measures overall correctness.

\[
\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}
\]

### Pros:
- Simple and intuitive

### Cons:
- Misleading for imbalanced datasets

---

## 4. Precision
Measures correctness of positive predictions.

\[
\text{Precision} = \frac{TP}{TP + FP}
\]

### Interpretation:
- High precision → few false positives

---

## 5. Recall (Sensitivity / True Positive Rate)
Measures ability to detect positive instances.

\[
\text{Recall} = \frac{TP}{TP + FN}
\]

### Interpretation:
- High recall → few false negatives

---

## 6. Specificity (True Negative Rate)
Measures ability to detect negative instances.

\[
\text{Specificity} = \frac{TN}{TN + FP}
\]

---

## 7. F1 Score
Harmonic mean of precision and recall.

\[
\text{F1} = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}
\]

### Why Harmonic Mean?
- Penalizes extreme values
- Requires both precision and recall to be high

---

## 8. Fβ Score
Generalized F-score.

\[
F_\beta = (1 + \beta^2) \cdot \frac{P \cdot R}{(\beta^2 \cdot P) + R}
\]

- `β > 1` → recall emphasized
- `β < 1` → precision emphasized

---

## 9. Error Rate
\[
\text{Error Rate} = 1 - \text{Accuracy}
\]

---

## 10. Probability-Based Metrics

### 10.1 Log Loss (Cross-Entropy Loss)
Evaluates predicted probabilities.

\[
\text{Log Loss} = -\frac{1}{N} \sum
\left[
y \log(\hat{y}) + (1 - y)\log(1 - \hat{y})
\right]
\]

- Lower is better
- Penalizes confident wrong predictions heavily

---

### 10.2 Brier Score
Mean squared error of probability predictions.

\[
\text{Brier} = \frac{1}{N} \sum (\hat{y} - y)^2
\]

---

## 11. ROC Curve (Receiver Operating Characteristic)

Plots:
- **True Positive Rate (Recall)** vs
- **False Positive Rate**

\[
\text{FPR} = \frac{FP}{FP + TN}
\]

### Characteristics:
- Threshold-independent
- Shows trade-off between sensitivity and specificity

---

## 12. AUC (Area Under ROC Curve)
\[
\text{AUC} \in [0, 1]
\]

### Interpretation:
- `0.5` → random guessing
- `1.0` → perfect classifier
- Higher AUC → better ranking ability

---

## 13. Precision–Recall (PR) Curve
Plots:
- Precision vs Recall

### Preferred when:
- Dataset is highly imbalanced
- Positive class is more important

---

## 14. Average Precision (AP)
Area under the Precision–Recall curve.

---

## 15. Threshold Selection
Classification depends on probability threshold.

Default:
\[
\hat{y} = 1 \text{ if } P(y=1|x) \ge 0.5
\]

Adjust threshold to:
- Increase recall → lower threshold
- Increase precision → higher threshold

---

## 16. Multiclass Classification Metrics

### 16.1 Confusion Matrix (Multiclass)
An `N × N` matrix where `N` is number of classes.

---

### 16.2 Macro Averaging
- Unweighted mean across classes
- Treats all classes equally

---

### 16.3 Micro Averaging
- Aggregates global TP, FP, FN
- Favors majority class

---

### 16.4 Weighted Averaging
- Weighted by class support

---

## 17. Cohen’s Kappa
Measures agreement beyond chance.

\[
\kappa = \frac{p_o - p_e}{1 - p_e}
\]

- `1` → perfect agreement
- `0` → chance level

---

## 18. Matthews Correlation Coefficient (MCC)
Balanced metric for binary classification.

\[
\text{MCC} = \frac{TP \cdot TN - FP \cdot FN}
{\sqrt{(TP+FP)(TP+FN)(TN+FP)(TN+FN)}}
\]

- Range: `[-1, 1]`
- Robust to class imbalance

---

## 19. Cost-Sensitive Metrics
In some applications:
- False Positives and False Negatives have different costs

Example:
\[
\text{Total Cost} = c_{FP} \cdot FP + c_{FN} \cdot FN
\]

---

## 20. Implementation (Scikit-Learn)

```python
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, roc_auc_score,
    log_loss, classification_report
)

accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)
cm = confusion_matrix(y_true, y_pred)
auc = roc_auc_score(y_true, y_prob)
loss = log_loss(y_true, y_prob)


| Scenario                | Recommended Metric    |
| ----------------------- | --------------------- |
| Balanced classes        | Accuracy              |
| Imbalanced data         | Precision, Recall, F1 |
| Ranking quality         | AUC                   |
| Probability calibration | Log Loss              |
| Medical diagnosis       | Recall                |
| Spam detection          | Precision             |


# Softmax Function — Notes for Classification

## 1. Introduction
The **Softmax function** is a generalization of logistic (sigmoid) regression used for **multiclass classification** problems. It converts a vector of raw scores (logits) into a **probability distribution** over multiple classes.

It is commonly used in:
- Multiclass Logistic Regression
- Neural Networks (final layer)

---

## 2. Mathematical Definition
Given a vector of logits:
\[
\mathbf{z} = [z_1, z_2, \dots, z_K]
\]

The softmax function for class `i` is defined as:
\[
\text{Softmax}(z_i) = \frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}}
\]

Where:
- `K` → number of classes
- `z_i` → raw score for class `i`

---

## 3. Key Properties
- Output values lie in **(0, 1)**
- Sum of probabilities equals **1**
\[
\sum_{i=1}^{K} \text{Softmax}(z_i) = 1
\]
- Preserves relative ordering of logits
- Differentiable (important for backpropagation)

---

## 4. Softmax vs Sigmoid

| Aspect | Sigmoid | Softmax |
|------|--------|---------|
| Use case | Binary classification | Multiclass classification |
| Output | Single probability | Probability vector |
| Sum to 1 | ❌ | ✅ |
| Decision | Threshold (0.5) | Argmax |

---

## 5. Softmax in Multiclass Logistic Regression

### Model
\[
z_k = w_k^T x + b_k
\]

\[
P(y = k | x) = \frac{e^{w_k^T x + b_k}}{\sum_{j=1}^{K} e^{w_j^T x + b_j}}
\]

Each class has:
- Its own weight vector `w_k`
- Its own bias `b_k`

---

## 6. Decision Rule
\[
\hat{y} = \arg\max_k P(y = k | x)
\]

The class with the highest probability is selected.

---

## 7. Numerical Stability
Direct computation may cause overflow due to large exponentials.

### Stable Softmax
\[
\text{Softmax}(z_i) =
\frac{e^{z_i - \max(z)}}{\sum_{j=1}^{K} e^{z_j - \max(z)}}
\]

- Subtracting `max(z)` does **not** change probabilities
- Prevents overflow

---

## 8. Loss Function: Categorical Cross-Entropy
Used with softmax:

\[
L = -\sum_{k=1}^{K} y_k \log(\hat{y}_k)
\]

For one-hot encoded labels:
- Only the true class contributes to loss

---

## 9. Gradient Intuition
The gradient of softmax combined with cross-entropy simplifies to:

\[
\frac{\partial L}{\partial z_k} = \hat{y}_k - y_k
\]

This simplicity makes training efficient.

---

## 10. Feature Handling
- Each feature contributes to **every class score**
- Weights represent how strongly a feature supports a class
- Linear decision boundaries between classes

---

## 11. Regularization
Same as logistic regression:
- L2 (Ridge): smooth decision boundaries
- L1 (Lasso): feature selection

---

## 12. Implementation (From Scratch)

```python
import numpy as np

def softmax(z):
    z = z - np.max(z, axis=1, keepdims=True)
    exp_z = np.exp(z)
    return exp_z / np.sum(exp_z, axis=1, keepdims=True)


| Metric                   | Range   | Best Used For                                         | Formula                                                                             |
| ------------------------ | ------- | ----------------------------------------------------- | ----------------------------------------------------------------------------------- |
| **Accuracy**             | [0, 1]  | Balanced datasets                                     | ( \frac{TP + TN}{TP + TN + FP + FN} )                                               |
| **Precision**            | [0, 1]  | Minimizing false positives                            | ( \frac{TP}{TP + FP} )                                                              |
| **Recall (Sensitivity)** | [0, 1]  | Minimizing false negatives                            | ( \frac{TP}{TP + FN} )                                                              |
| **F1 Score**             | [0, 1]  | Balance between precision and recall                  | ( 2 \times \frac{Precision \times Recall}{Precision + Recall} )                     |
| **Specificity**          | [0, 1]  | Minimizing false positives                            | ( \frac{TN}{TN + FP} )                                                              |
| **Log Loss**             | [0, ∞)  | Penalizing incorrect predictions with high confidence | ( - \sum \left[ y_i \log(p_i) + (1 - y_i) \log(1 - p_i) \right] )                   |
| **MCC**                  | [-1, 1] | Balanced metric for imbalanced classes                | ( \frac{TP \times TN - FP \times FN}{\sqrt{(TP + FP)(TP + FN)(TN + FP)(TN + FN)}} ) |
| **Cohen's Kappa**        | [-1, 1] | Agreement between raters or classifiers               | ( \frac{p_o - p_e}{1 - p_e} )                                                       |
| **AUC-PR**               | [0, 1]  | Imbalanced datasets and focusing on positive class    | N/A                                                                                 |


### ROC Curve and AUC Score: Explanation, Definitions, and Examples

#### **1. ROC Curve (Receiver Operating Characteristic Curve)**

**Definition**:
The ROC curve is a graphical representation of the performance of a binary classifier system as its discrimination threshold is varied. It shows the relationship between **True Positive Rate (TPR)** and **False Positive Rate (FPR)**.

* **True Positive Rate (TPR)**: Also known as sensitivity or recall, this is the ratio of correctly predicted positive instances to all actual positive instances.

  [
  TPR = \frac{TP}{TP + FN}
  ]
  where:

  * TP = True Positives
  * FN = False Negatives

* **False Positive Rate (FPR)**: This is the ratio of incorrectly predicted positive instances to all actual negative instances.

  [
  FPR = \frac{FP}{FP + TN}
  ]
  where:

  * FP = False Positives
  * TN = True Negatives

**ROC Curve**:
The ROC curve plots the TPR (on the Y-axis) against the FPR (on the X-axis) for different threshold values. As you adjust the threshold for predicting positive versus negative cases, the points on the ROC curve change.

* **A perfect classifier**: The ROC curve will touch the top-left corner, meaning a high true positive rate and a low false positive rate.
* **A random classifier**: The curve will lie along the diagonal line, indicating no discrimination ability between classes.

**Example**:
Imagine a classifier predicting whether an email is spam or not. For a certain threshold, the classifier might classify some non-spam emails as spam (False Positives) and some spam emails as non-spam (False Negatives). As you adjust the threshold, the ROC curve shows how the TPR and FPR change.

---

#### **2. AUC (Area Under the Curve)**

**Definition**:
AUC is the area under the ROC curve and serves as a summary statistic for the ROC curve. It represents the probability that the model will rank a randomly chosen positive instance higher than a randomly chosen negative instance. AUC provides an aggregate measure of the classifier's ability to discriminate between the positive and negative classes.

* **AUC value range**: [0, 1]

  * **AUC = 1**: Perfect model (perfect classification).
  * **AUC = 0.5**: The model performs no better than random chance (similar to flipping a coin).
  * **AUC < 0.5**: The model performs worse than random chance (indicating the model is predicting the wrong class more often).

**AUC Interpretation**:

* **AUC = 0.90-1.00**: Excellent model (very high discrimination ability).
* **AUC = 0.80-0.90**: Good model.
* **AUC = 0.70-0.80**: Fair model.
* **AUC = 0.60-0.70**: Poor model.
* **AUC = 0.50-0.60**: No discrimination ability, random guessing.

**Example**:
If you have a model predicting whether a patient has a certain disease (positive class) or not (negative class), an AUC of 0.85 means there is an 85% chance that the model will correctly rank a randomly selected diseased patient higher than a healthy patient.

---

### **3. Understanding ROC Curve and AUC in Practice**

#### **Step-by-Step Example**

Let's walk through an example using a classifier with the following results:

| Actual / Predicted    | Positive (Predicted) | Negative (Predicted) |
| --------------------- | -------------------- | -------------------- |
| **Positive (Actual)** | 50 (True Positives)  | 10 (False Negatives) |
| **Negative (Actual)** | 5 (False Positives)  | 100 (True Negatives) |

Here’s how to calculate key metrics:

1. **True Positive Rate (TPR)**:
   [
   TPR = \frac{TP}{TP + FN} = \frac{50}{50 + 10} = \frac{50}{60} = 0.833
   ]

2. **False Positive Rate (FPR)**:
   [
   FPR = \frac{FP}{FP + TN} = \frac{5}{5 + 100} = \frac{5}{105} \approx 0.048
   ]

The ROC curve will plot these values for different thresholds. For each threshold, the number of True Positives and False Positives changes, and the TPR and FPR are recalculated.

---

#### **4. When to Use ROC Curve and AUC**

* **Imbalanced Data**: ROC and AUC are useful for binary classification problems with imbalanced datasets, where the number of instances in one class is much higher than the other. The AUC value gives a more balanced measure of performance compared to accuracy, which may be misleading in such cases.

* **Comparing Models**: ROC and AUC are excellent for comparing different models. A model with a higher AUC is typically preferred over a model with a lower AUC because it indicates better overall discrimination.

---

### **5. Limitations**

* **Threshold Dependency**: While the ROC curve is helpful, it does not tell you the optimal threshold for classification. This is often addressed by plotting Precision-Recall curves when the dataset is highly imbalanced.

* **Not always informative for imbalanced datasets**: In some cases (especially with highly imbalanced data), AUC can be overly optimistic. In such cases, precision, recall, and the F1 score may be more meaningful.

---

### **6. Practical Use in Python (Scikit-learn)**

You can calculate and plot the ROC curve and AUC using libraries like Scikit-learn. Here's a simple example:

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
from sklearn.ensemble import RandomForestClassifier

# Create a synthetic dataset
X, y = make_classification(n_samples=1000, n_features=20, n_classes=2, random_state=42)

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train a classifier
classifier = RandomForestClassifier()
classifier.fit(X_train, y_train)

# Predict probabilities
y_probs = classifier.predict_proba(X_test)[:, 1]  # Probabilities for the positive class

# Compute ROC curve and AUC
fpr, tpr, thresholds = roc_curve(y_test, y_probs)
roc_auc = auc(fpr, tpr)

# Plot ROC curve
plt.figure()
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC)')
plt.legend(loc='lower right')
plt.show()
```

Sure! Apart from **ROC** and **AUC**, there are several other metrics commonly used to evaluate the performance of classification models. These metrics can provide a more comprehensive view of how well a model is performing, especially in different situations like imbalanced datasets, varying costs of false positives/negatives, and more.

Here's a rundown of the most commonly used classification metrics, including definitions, formulas, and when to use them:

---

### 1. **Accuracy**

**Definition**:
Accuracy is the most straightforward metric, representing the proportion of correct predictions (both True Positives and True Negatives) out of all predictions.

**Formula**:
[
\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}
]
Where:

* **TP** = True Positives
* **TN** = True Negatives
* **FP** = False Positives
* **FN** = False Negatives

**Use Case**:
Accuracy is a good metric when the classes are balanced. However, in imbalanced datasets, accuracy can be misleading, as a classifier can achieve high accuracy by predicting the majority class most of the time.

**Example**:
For a dataset of 1000 emails, where 950 are non-spam (negative class) and 50 are spam (positive class), predicting all emails as non-spam would give 95% accuracy but would miss all spam emails.

---

### 2. **Precision (Positive Predictive Value)**

**Definition**:
Precision measures the proportion of positive predictions that were actually correct. In other words, it focuses on the accuracy of the positive class predictions.

**Formula**:
[
\text{Precision} = \frac{TP}{TP + FP}
]

**Use Case**:
Precision is important when the cost of false positives is high. For example, in medical testing, if a test incorrectly identifies healthy patients as sick (false positives), it might lead to unnecessary treatments or interventions.

**Example**:
In a cancer screening test, if the test wrongly classifies 10 healthy people as having cancer, the **precision** will be low because a high proportion of the predicted positives are incorrect.

---

### 3. **Recall (Sensitivity, True Positive Rate)**

**Definition**:
Recall measures the proportion of actual positives that were correctly identified. It answers the question: *Of all the actual positive cases, how many did we catch?*

**Formula**:
[
\text{Recall} = \frac{TP}{TP + FN}
]

**Use Case**:
Recall is important when missing a positive instance is costly. For instance, in fraud detection, it's critical to identify as many fraudulent transactions as possible, even if that means some legitimate transactions are flagged.

**Example**:
In a disease detection test, recall would be high if the model correctly identifies most of the sick patients, even if it also incorrectly classifies some healthy patients as sick (false positives).

---

### 4. **F1 Score (F-Score)**

**Definition**:
The F1 score is the harmonic mean of **precision** and **recall**. It balances the trade-off between precision and recall, particularly useful when the class distribution is imbalanced.

**Formula**:
[
\text{F1} = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}
]

**Use Case**:
F1 score is preferred when you need a balance between precision and recall and when you care about both false positives and false negatives equally.

**Example**:
In a spam detection system, you might care about both identifying as many spam emails as possible (high recall) and ensuring that you don't classify too many legitimate emails as spam (high precision). The **F1 score** provides a single metric to optimize both aspects.

---

### 5. **Specificity (True Negative Rate)**

**Definition**:
Specificity measures the proportion of actual negatives that are correctly identified. It answers the question: *Of all the actual negative cases, how many did we catch?*

**Formula**:
[
\text{Specificity} = \frac{TN}{TN + FP}
]

**Use Case**:
Specificity is important when the cost of false positives is high. For instance, in security systems, if an alarm goes off too often (false positives), it may become ignored.

**Example**:
In a medical test for a rare disease, specificity would measure how many healthy people were correctly identified as healthy.

---

### 6. **Logarithmic Loss (Log Loss)**

**Definition**:
Logarithmic loss (or **log loss**) quantifies the uncertainty of your model’s predictions. It measures the "distance" between the predicted probability and the actual class label. The lower the log loss, the better the model.

**Formula**:
[
\text{Log Loss} = -\frac{1}{N} \sum_{i=1}^{N} \left[ y_i \log(p_i) + (1 - y_i) \log(1 - p_i) \right]
]
Where:

* **y_i** = Actual class label for the ith instance (0 or 1)
* **p_i** = Predicted probability of the positive class for the ith instance
* **N** = Total number of instances

**Use Case**:
Log loss is mainly used for **probabilistic models** (e.g., logistic regression, neural networks) and is useful when you want to penalize incorrect confidence more severely.

**Example**:
If a model predicts that a patient has a disease with 0.9 probability, but the patient is healthy, this will result in a high log loss, meaning the model was very confident, but wrong.

---

### 7. **Matthews Correlation Coefficient (MCC)**

**Definition**:
The Matthews Correlation Coefficient is a balanced metric that takes into account all four confusion matrix categories (TP, TN, FP, FN). It ranges from -1 (perfectly wrong) to +1 (perfectly right), with 0 indicating random predictions.

**Formula**:
[
\text{MCC} = \frac{TP \times TN - FP \times FN}{\sqrt{(TP + FP)(TP + FN)(TN + FP)(TN + FN)}}
]

**Use Case**:
MCC is especially useful when dealing with imbalanced datasets, as it considers both classes equally and is not biased by class distribution.

**Example**:
In a rare event prediction (e.g., predicting fraud or disease), MCC gives a better understanding of model performance across all classes.

---

### 8. **Cohen's Kappa**

**Definition**:
Cohen's Kappa measures the agreement between two raters or classifiers, adjusted for chance. It ranges from -1 (complete disagreement) to 1 (complete agreement), with 0 indicating no better agreement than random chance.

**Formula**:
[
\text{Kappa} = \frac{p_o - p_e}{1 - p_e}
]
Where:

* ( p_o ) = Observed agreement
* ( p_e ) = Expected agreement (random agreement)

**Use Case**:
Cohen's Kappa is used when you need to evaluate the consistency of a classifier against a baseline (or another classifier) while correcting for the possibility of agreement occurring by chance.

---

### 9. **Area Under the Precision-Recall Curve (AUC-PR)**

**Definition**:
The **Precision-Recall (PR) curve** is another way to evaluate binary classifiers, especially when dealing with imbalanced classes. AUC-PR is the area under this curve, and it summarizes the trade-off between **precision** and **recall** at different thresholds.

**Use Case**:
AUC-PR is particularly useful when you have a highly imbalanced dataset. Unlike AUC-ROC, which can be overly optimistic in such cases, AUC-PR focuses only on the positive class.
