# Equation of a Line, 3D, HyperPlane

Sure — here’s a **more detailed, structured, exam-ready explanation**, starting from basics and building up to higher dimensions.

---

**1. Equation of a Line (2D)**

**a) Slope of a Line**

The **slope** (m) of a line passing through two points
((x_1, y_1)) and ((x_2, y_2)) is:

[
m = \frac{y_2 - y_1}{x_2 - x_1}
]

---

**b) Point–Slope Form**

If a line has slope (m) and passes through point ((x_1, y_1)):

[
y - y_1 = m(x - x_1)
]

---

**c) Slope–Intercept Form**

If the y-intercept is (c):

[
y = mx + c
]

---

**d) Two-Point Form**

Line passing through two points ((x_1,y_1)) and ((x_2,y_2)):

[
\frac{y - y_1}{y_2 - y_1} = \frac{x - x_1}{x_2 - x_1}
]

---

**e) General Form**

[
Ax + By + C = 0
]

Slope from general form:
[
m = -\frac{A}{B}
]

---
**2. Equation of a Line in 3D**

**a) Direction Ratios**

If a line has direction ratios ((a,b,c)) and passes through
((x_0,y_0,z_0)):

---

**b) Vector Form**

[
\vec{r} = \vec{r_0} + t\vec{v}
]

---

**c) Parametric Form**

[
\begin{aligned}
x &= x_0 + at \
y &= y_0 + bt \
z &= z_0 + ct
\end{aligned}
]

---
**d) Symmetric Form**

[
\frac{x - x_0}{a} = \frac{y - y_0}{b} = \frac{z - z_0}{c}
]

---

**3. Equation of a Plane (3D)**

**a) General Form**

[
Ax + By + Cz + D = 0
]

Here:

* ((A,B,C)) is the **normal vector** to the plane

---

**b) Point–Normal Form**

Plane passing through point ((x_0,y_0,z_0)) with normal ((A,B,C)):

[
A(x - x_0) + B(y - y_0) + C(z - z_0) = 0
]

---

**c) Intercept Form**

If plane cuts axes at (a,b,c):

[
\frac{x}{a} + \frac{y}{b} + \frac{z}{c} = 1
]

---

**4. Equation of a Hyperplane (n-D)**

A **hyperplane** is the generalization of a line and plane to (n) dimensions.

### **a) Standard Form**

[
\mathbf{w} \cdot \mathbf{x} + b = 0
]

or

[
\sum_{i=1}^{n} w_i x_i + b = 0
]

---

**b) Expanded Example (4D)**

[
w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4 + b = 0
]

---

**c) ML Interpretation**

* (\mathbf{w}): weights (normal vector)
* (b): bias
* Used in **Perceptron, Logistic Regression, SVM**
