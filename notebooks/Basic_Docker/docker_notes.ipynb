# Docker
## 🐳 What is Docker? (Top → Bottom Overview)

Docker is an open-source platform used to build, package, ship, and run applications inside containers.

A container is a lightweight package that includes:

1. Application code
2. Runtime
3. Libraries
4. Dependencies
5. System tools

This ensures the application runs the same everywhere (developer laptop → testing → production).

**Why Docker Exists**

Before Docker:

Apps ran differently across machines.

Developers said “It works on my machine.”

Docker solves this by packaging everything the app needs.

Containers vs Virtual Machines
| Feature      | Containers    | Virtual Machines |
| ------------ | ------------- | ---------------- |
| Startup time | Seconds       | Minutes          |
| Size         | MBs           | GBs              |
| Isolation    | Process-level | Full OS          |
| Performance  | Near-native   | Slight overhead  |


Docker containers share the host OS kernel, making them much lighter than VMs.

🧱 Docker Architecture

Main components of Docker:

1. Docker Client – CLI where you run commands (docker run, docker build)
2. Docker Daemon – Background service managing containers
3. Docker Images – Read-only templates for containers
4. Docker Containers – Running instances of images
5. Docker Registry – Storage for images (like Docker Hub)

Workflow:
Dockerfile → Image → Container → Run Application

### **📦 Core Docker Concepts**

### 1️ Docker Image

A **blueprint** for containers.

Example:

```
python:3.11
node:18
ubuntu:22.04
```

Images are layered and immutable.

---

### 2️ Docker Container

A **running instance of an image**.

Example:

```
docker run nginx
```

---

### 3️ Dockerfile

A **script used to build Docker images**.

Example:

```dockerfile
FROM node:18
WORKDIR /app
COPY . .
RUN npm install
CMD ["node","server.js"]
```

---

### 4️ Docker Registry

Stores images online.

Example registries:

* **Docker Hub**
* **GitHub Container Registry**
* **Amazon Elastic Container Registry**

### ⚙️ Most Important Docker Commands

#### 1️ Docker Installation Check

```bash
docker --version
```
Shows installed Docker version.

---

#### 📦 Image Commands

##### Pull Image

```bash
docker pull nginx
```

Downloads image from Docker Hub.

---

#### List Images

```bash
docker images
```

Shows downloaded images.

---

#### Remove Image

```bash
docker rmi image_id
```

Example:

```bash
docker rmi nginx
```

---

#### 🟢 Container Commands

##### Run Container

```bash
docker run nginx
```

Run container from image.

Common flags:

```bash
-d   # run in background
-p   # port mapping
--name container_name
```

Example:

```bash
docker run -d -p 8080:80 --name web nginx
```

---

#### List Running Containers

```bash
docker ps
```

All containers:

```bash
docker ps -a
```

---

#### Stop Container

```bash
docker stop container_id
```

---

#### Start Container

```bash
docker start container_id
```

---

#### Remove Container

```bash
docker rm container_id
```

---

### 🛠️ Container Debugging Commands

#### Container Logs

```bash
docker logs container_id
```

---

#### Execute Command Inside Container

```bash
docker exec -it container_id bash
```

Example:

```bash
docker exec -it web bash
```

---

#### Inspect Container

```bash
docker inspect container_id
```

Shows JSON details.

---

### 🏗️ Build & Development Commands

#### Build Image

```bash
docker build -t myapp .
```

`-t` = tag name.

---

#### Tag Image

```bash
docker tag image_name username/image_name
```

---

#### Push Image

```bash
docker push username/image_name
```

Uploads to Docker Hub.

---

### 📂 Volume Commands (Data Persistence)

Create volume:

```bash
docker volume create myvolume
```

Run container with volume:

```bash
docker run -v myvolume:/data nginx
```

List volumes:

```bash
docker volume ls
```

---

### 🌐 Network Commands

List networks:

```bash
docker network ls
```

Create network:

```bash
docker network create mynetwork
```

Run container in network:

```bash
docker run --network=mynetwork nginx
```

---

### 📦 Docker Compose (Multi-Container Apps)

Docker Compose lets you run **multiple containers together**.

Example `docker-compose.yml`:

```yaml
version: '3'
services:
  web:
    image: nginx
    ports:
      - "8080:80"
  db:
    image: mysql
```

Run:

```bash
docker compose up
```

Stop:

```bash
docker compose down
```
