```python
# Create a NumPy array of shape (3, 3) with values from 1 to 9
array = np.arange(1, 10).reshape((3, 3))
print("Original array:")
print(array)

# Normalize the array
mean = np.mean(array)
std_dev = np.std(array)
normalized_array = (array - mean) / std_dev

print("Mean:")
print(mean)

print("std_dev:")
print(std_dev)

print("Normalized array:")
print(normalized_array)
```

```python
# Create a NumPy array of shape (3, 3) filled with random integers
array = np.random.randint(1, 11, size=(3, 3))
row_array = np.random.randint(1, 11, size=(3,))
print("Original array:")
print(array)
print("1D array:")
print(row_array)

# Add the 1D array to each row of the 2D array using broadcasting
result = array + row_array
print("Resulting array:")
print(result)
```

```python
# Create a NumPy array of shape (4, 4) filled with random integers
array = np.random.randint(1, 11, size=(4, 4))
column_array = np.random.randint(1, 11, size=(4,))
print("Original array:")
print(array)
print("1D array:")
print(column_array)

# Subtract the 1D array from each column of the 2D array using broadcasting
result = array - column_array[:, np.newaxis]
print("Resulting array:")
print(result)
```

```python
# Create a structured array with fields 'x' and 'y'
data_type = [('x', 'i4'), ('y', 'i4')]
data = np.array([(1, 2), (3, 4), (5, 6)], dtype=data_type)
print("Original array:")
print(data)

# Compute the Euclidean distance between each pair of points
#distance(i, j) = √[(xᵢ - xⱼ)² + (yᵢ - yⱼ)²]
distances = np.sqrt((data['x'][:, np.newaxis] - data['x'])**2 + (data['y'][:, np.newaxis] - data['y'])**2)
print("Euclidean distances:")
print(distances)
print(distances.dtype)
```

distance(i, j) = √[(xᵢ - xⱼ)² + (yᵢ - yⱼ)²]

This produces a 3x3 matrix like:

[[d(0,0) d(0,1) d(0,2)]
 [d(1,0) d(1,1) d(1,2)]
 [d(2,0) d(2,1) d(2,2)]]

🔍 Breaking down the distance line:
(data['x'][:, np.newaxis] - data['x'])**2


Let’s take it piece by piece.

data['x']:

This is just:

[1 3 5]

data['x'][:, np.newaxis]:

Adds a new axis to make it a column vector:

[[1]
 [3]
 [5]]


So now you can broadcast this against the row vector [1 3 5] to get all pairwise x-differences.

(data['x'][:, np.newaxis] - data['x']):

This gives you the matrix of xᵢ - xⱼ:

[[ 0 -2 -4]
 [ 2  0 -2]
 [ 4  2  0]]


Now square it:

[[ 0  4 16]
 [ 4  0  4]
 [16  4  0]]


Do the same for data['y'], you’ll get:

[[ 0  4 16]
 [ 4  0  4]
 [16  4  0]]

Add the squared differences:
(x_diff_squared) + (y_diff_squared)

[[ 0  8 32]
 [ 8  0  8]
 [32  8  0]]

Take the square root:
np.sqrt(...)

[[0.         2.82842712 5.65685425]
 [2.82842712 0.         2.82842712]
 [5.65685425 2.82842712 0.        ]]

```python
data = np.random.randint(1,6,10)
print(data)
# print(map(set(data),data+1))


counts = pd.Series(data).value_counts().sort_index()
print(counts)
plt.hist(data,bins=[1,2,3,4,5],color='blue',edgecolor='white')
plt.ylabel('count')
plt.xlabel('values')

# Convert counts to table data
table_data = [[val] for val in counts.values]
row_labels = [str(i) for i in counts.index]
col_labels = ['Count']

# Add table below the histogram
table = plt.table(
    cellText=table_data,
    rowLabels=row_labels,
    colLabels=col_labels,
    cellLoc='center',
    loc='bottom'
)

table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1, 1.2)

plt.tight_layout()
plt.show()
```

```python
data = np.random.randint(1, 6, 20)
counts = pd.Series(data).value_counts().sort_index()

# Prepare table data
table_data = [[val] for val in counts.values]
# counts is a Pandas Series that holds the frequency of each value (1–5) in your dataset.
# counts.values gives you just the frequencies, e.g., [3, 4, 5, 6, 2] (these are example frequencies).
# We wrap each value in a list: [[3], [4], [5], [6], [2]]
# Why? Because plt.table() expects cellText to be a 2D list, where each sublist represents a row in the table.
row_labels = [str(i) for i in counts.index]
# counts.index are the unique values from the original data, e.g., [1, 2, 3, 4, 5]
# str(i) converts the index (which are integers) to strings, as required for table labels.
# So row_labels becomes: ['1', '2', '3', '4', '5']
col_labels = ['Count']

# Create two subplots: one for histogram, one for table
fig, (ax1, ax2) = plt.subplots(
    2, 1,
    figsize=(6, 6),
    gridspec_kw={'height_ratios': [4, 1]}
)

# --- Plot histogram ---
ax1.hist(data, bins=4, color='blue', edgecolor='white')
ax1.set_ylabel('Count')
ax1.set_xlabel('Values')
ax1.set_title('Histogram with Value Count Table')

# --- Plot table ---
ax2.axis('off')  # Hide the actual subplot axes
table = ax2.table(
    cellText=table_data,
    rowLabels=row_labels,
    colLabels=col_labels,
    loc='center',
    cellLoc='center'
)
table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1, 1.2)  # Adjust size of table cells

# Tidy layout
plt.tight_layout()
plt.show()
```

```python
# Data
labels = ['Watermelon', 'Orange', 'Cherries', 'Mango']
sizes = [25, 30, 20, 25]
colors = ['red', 'orange', 'lightcoral', 'yellow']

# Dynamically explode only the max value
max_index = sizes.index(max(sizes))
explode = [0.1 if i == max_index else 0 for i in range(len(sizes))]

# Plot
plt.pie(
    sizes,
    labels=labels,
    explode=explode,
    colors=colors,
    autopct='%1.1f%%',
    shadow=True,
    startangle=90
)
plt.title('Fruit Distribution')
plt.axis('equal')

plt.show()
```

```python

labels = ['Watermelon', 'Orange', 'Cherries', 'Mango']
sizes = [23, 30, 20, 27]
colors = ['red', 'orange', 'lightcoral', 'yellow']

# Explode slices where size >= 25
explode = [0.1 if size > 25 else 0 for size in sizes]

plt.pie(
    sizes,
    labels=labels,
    explode=explode,
    colors=colors,
    autopct='%1.1f%%',
    shadow=True,
    startangle=90
)
plt.title('Fruit Distribution')
plt.axis('equal')
plt.show()
```
