```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import random

# Simulated drone positions
drone_positions = {
    'drone1': [0, 0, 5],
    'drone2': [2, 2, 6],
    'drone3': [-1, 3, 4],
}

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Update loop
for i in range(100):  # simulate 100 time steps
    ax.clear()
    ax.set_xlim(-10, 10)
    ax.set_ylim(-10, 10)
    ax.set_zlim(0, 10)
    ax.set_title("Swarm Drone 3D Visualization")

    for drone_id, pos in drone_positions.items():
        # Simulate random movement
        pos[0] += random.uniform(-0.2, 0.2)
        pos[1] += random.uniform(-0.2, 0.2)
        pos[2] += random.uniform(-0.1, 0.1)

        ax.scatter(pos[0], pos[1], pos[2], label=drone_id)

    plt.legend()
    plt.pause(0.1)
    
```

```python
import graphviz

def generate_ec_flowchart():
    # Create a new directed graph
    dot = graphviz.Digraph('EC_Process', comment='Generalized EC Process Flowchart')
    dot.attr(rankdir='TB', size='8,10')

    # Define Node Styles
    dot.attr('node', shape='ellipse', style='filled', color='lightblue')
    dot.node('start', 'Initialize\n(Create Initial Population)')
    dot.node('output', 'Output\n(Best Solution)')

    dot.attr('node', shape='rect', style='filled', color='lightgrey')
    dot.node('eval', 'Evaluate\n(Assess Fitness)')
    dot.node('select', 'Select\n(Apply Selection Pressure)')
    dot.node('reproduce', 'Reproduce\n(Mutation & Crossover)')
    dot.node('replace', 'Replace\n(Update Population)')

    dot.attr('node', shape='diamond', style='filled', color='salmon')
    dot.node('term', 'Terminate?')

    # Define Edges (The Flow)
    dot.edge('start', 'eval')
    dot.edge('eval', 'term')
    
    # Decision branches
    dot.edge('term', 'output', label='Yes')
    dot.edge('term', 'select', label='No')
    
    # The Loop
    dot.edge('select', 'reproduce')
    dot.edge('reproduce', 'replace')
    dot.edge('replace', 'eval')

    # Render the graph
    dot.render('ec_flowchart', format='png', cleanup=True)
    print("Flowchart generated as 'ec_flowchart.png'")
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, Rectangle, Polygon, FancyArrowPatch

# ---------- helpers ----------
def rounded_box(ax, x, y, w, h, text, fontsize=12, lw=1.8):
    box = FancyBboxPatch(
        (x, y), w, h,
        boxstyle="round,pad=0.02,rounding_size=0.04",
        linewidth=lw, edgecolor="black", facecolor="white"
    )
    ax.add_patch(box)
    ax.text(x + w/2, y + h/2, text, ha="center", va="center",
            fontsize=fontsize, fontweight="bold")
    return box

def rect_box(ax, x, y, w, h, text, fontsize=12, lw=1.8):
    box = Rectangle((x, y), w, h, linewidth=lw,
                    edgecolor="black", facecolor="white")
    ax.add_patch(box)
    ax.text(x + w/2, y + h/2, text, ha="center", va="center",
            fontsize=fontsize, fontweight="bold")
    return box

def diamond(ax, cx, cy, w, h, text, fontsize=12, lw=1.8):
    pts = [
        (cx, cy + h/2),
        (cx + w/2, cy),
        (cx, cy - h/2),
        (cx - w/2, cy),
    ]
    poly = Polygon(pts, closed=True, linewidth=lw,
                   edgecolor="black", facecolor="white")
    ax.add_patch(poly)
    ax.text(cx, cy, text, ha="center", va="center",
            fontsize=fontsize, fontweight="bold")
    return poly

def arrow(ax, xy1, xy2, text=None, text_xy=None, fontsize=11, connectionstyle="arc3,rad=0.0"):
    arr = FancyArrowPatch(
        xy1, xy2, arrowstyle='-|>', mutation_scale=16,
        linewidth=1.6, color='black', connectionstyle=connectionstyle
    )
    ax.add_patch(arr)
    if text and text_xy:
        ax.text(text_xy[0], text_xy[1], text, fontsize=fontsize, ha="center", va="center")
    return arr

# ---------- canvas ----------
fig, ax = plt.subplots(figsize=(16, 9))
ax.set_xlim(0, 16)
ax.set_ylim(0, 9)
ax.axis("off")

# ---------- title ----------
ax.text(8, 8.7, "Fig. 1. Generalized EC process",
        ha="center", va="center", fontsize=18, fontweight="bold")

# ---------- dashed generalized cycle boundary ----------
boundary = Rectangle((4.0, 1.0), 8.4, 6.9, fill=False,
                     linestyle="--", linewidth=1.5, edgecolor="black")
ax.add_patch(boundary)
ax.text(8.2, 7.45, "Generalized EC Cycle",
        ha="center", va="center", fontsize=15, fontweight="bold")

# ---------- left label ----------
ax.text(1.8, 3.3, "Generalized EC\nCycle (Engine of\nDirectional Improvement)",
        ha="center", va="center", fontsize=13, fontweight="bold")

# ---------- main blocks ----------
rounded_box(ax, 0.3, 4.55, 2.6, 1.1, "Initialize\n(Create Initial\nPopulation)", fontsize=12)
rect_box(ax,    5.0, 5.15, 2.4, 1.1, "Evaluate\n(Assess Fitness of\nIndividuals)", fontsize=12)
diamond(ax,     10.2, 5.65, 2.0, 1.6, "Terminate?\n(Condition met?)", fontsize=12)
rounded_box(ax, 13.0, 4.55, 2.5, 1.1, "Output\n(Best Solution Found)", fontsize=12)

rect_box(ax,    8.1, 3.05, 2.1, 1.1, "Select\n(Selection Pressure\napplied)", fontsize=12)
rect_box(ax,    8.1, 1.55, 2.1, 1.1, "Reproduce\n(Apply Mutation,\nCrossover)", fontsize=12)
rect_box(ax,    5.0, 2.75, 2.4, 1.1, "Replace\n(Update Population)", fontsize=12)

# ---------- arrows ----------
# Initialize -> Evaluate
arrow(ax, (2.9, 5.1), (5.0, 5.7))

# Evaluate -> Terminate
arrow(ax, (7.4, 5.7), (9.25, 5.7))

# Terminate -> Output (Yes)
arrow(ax, (11.2, 6.0), (13.0, 5.1), text="Yes", text_xy=(12.15, 5.6), fontsize=12)

# Terminate -> Select (No)
arrow(ax, (10.2, 4.85), (9.15, 4.15), text="No", text_xy=(9.85, 4.45), fontsize=12)

# Select -> Reproduce
arrow(ax, (9.15, 3.05), (9.15, 2.65))

# Reproduce -> Replace
arrow(ax, (8.1, 2.1), (7.4, 2.1))
arrow(ax, (7.4, 2.1), (7.4, 3.25))

# Replace -> Evaluate
arrow(ax, (6.2, 3.85), (6.2, 5.15))

# ---------- side annotations ----------
ax.text(14.1, 4.0, "Without selection\npressure, reduces to\nrandom sampling.",
        ha="left", va="center", fontsize=11)
ax.text(14.1, 1.9, "Without mutation,\nconverges to initial gene pool\nregardless of generations.",
        ha="left", va="center", fontsize=11)

# small connector lines to side notes
ax.plot([13.0, 12.7, 12.7], [4.15, 4.15, 4.0], color="black", linewidth=1.2)
ax.plot([10.2, 12.7, 12.7], [2.1, 2.1, 1.9], color="black", linewidth=1.2)

# ---------- save / show ----------
plt.tight_layout()
plt.show()
# plt.savefig("generalized_ec_flowchart.png", dpi=300, bbox_inches="tight")
```

```python
import matplotlib.pyplot as plt
import networkx as nx

def draw_networkx_flow():
    G = nx.DiGraph()
    
    # Define the nodes
    nodes = ["Initialize", "Evaluate", "Terminate?", "Select", "Reproduce", "Replace", "Output"]
    G.add_nodes_from(nodes)

    # FIX: Use nx.add_path(G, ...) instead of G.add_path(...)
    nx.add_path(G, ["Initialize", "Evaluate", "Terminate?"])
    G.add_edge("Terminate?", "Output")
    G.add_edge("Terminate?", "Select")
    nx.add_path(G, ["Select", "Reproduce", "Replace", "Evaluate"])

    # Manual positioning for a clean flowchart look
    pos = {
        "Initialize": (0, 5),
        "Evaluate": (0, 4),
        "Terminate?": (0, 3),
        "Output": (1.5, 3),
        "Select": (0, 2),
        "Reproduce": (-1.5, 2),
        "Replace": (-1.5, 4)
    }

    plt.figure(figsize=(10, 7))
    
    # Draw nodes
    nx.draw_networkx_nodes(G, pos, node_shape='s', node_size=4000, node_color='skyblue')
    
    # Draw labels
    nx.draw_networkx_labels(G, pos, font_size=10, font_weight='bold')
    
    # Draw edges with specific curvature for the loop
    nx.draw_networkx_edges(G, pos, edgelist=[("Initialize", "Evaluate"), ("Evaluate", "Terminate?"), 
                                            ("Terminate?", "Output"), ("Terminate?", "Select"),
                                            ("Select", "Reproduce"), ("Reproduce", "Replace")], 
                           arrowsize=20)
    
    # Add a curved edge for the "Replace -> Evaluate" feedback loop
    nx.draw_networkx_edges(G, pos, edgelist=[("Replace", "Evaluate")], 
                           arrowsize=20, connectionstyle="arc3,rad=0.1")

    plt.title("Fig 1. Generalized EC Process Flowchart", pad=20)
    plt.axis('off')
    plt.tight_layout()
    plt.show()

draw_networkx_flow()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch
import matplotlib.patheffects as pe

fig_h = 14
fig, ax = plt.subplots(figsize=(9, fig_h))
fig.patch.set_facecolor('#0D1B2A')
ax.set_facecolor('#0D1B2A')
ax.set_xlim(0, 10)
ax.set_ylim(0, fig_h)
ax.axis('off')

# Steps: (center_x, center_y, title, subtitle, color)
steps = [
    (5, 12.8, "INITIALIZATION",        "Define solution space & generate\nrandom population of size N",        "#1B4F72"),
    (5, 10.8, "FITNESS EVALUATION",    "Score each individual via\nfitness function  f(x) = 1/d(x)",         "#1A5276"),
    (5,  8.8, "SELECTION",             "Choose fittest parents\n(Roulette Wheel / Tournament)",               "#154360"),
    (5,  6.8, "REPRODUCTION",          "Crossover (p_c) + Mutation (p_m)\nto produce offspring",              "#0E6655"),
    (5,  4.8, "REPLACEMENT",           "Replace weaker individuals\nwith offspring (elitism / generational)", "#1B6B48"),
    (5,  2.8, "TERMINATION CHECK",     "Max generations reached\nor convergence condition met?",             "#6E2F1A"),
]

BOX_W = 6.2
BOX_H = 1.5

for (cx, cy, title, desc, color) in steps:
    x0 = cx - BOX_W / 2
    y0 = cy - BOX_H / 2
    box = FancyBboxPatch((x0, y0), BOX_W, BOX_H,
                          boxstyle="round,pad=0.12",
                          linewidth=1.4, edgecolor='#AAD4F5',
                          facecolor=color, zorder=3)
    ax.add_patch(box)
    ax.text(cx, cy + 0.3, title, ha='center', va='center',
            fontsize=11.5, fontweight='bold', color='white', zorder=4)
    ax.text(cx, cy - 0.28, desc, ha='center', va='center',
            fontsize=8.2, color='#D6EAF8', zorder=4, linespacing=1.4)

# Down arrows between steps
for i in range(len(steps) - 1):
    y_top = steps[i][1] - BOX_H / 2
    y_bot = steps[i+1][1] + BOX_H / 2
    mid_y = (y_top + y_bot) / 2
    ax.annotate("", xy=(5, y_bot + 0.04), xytext=(5, y_top - 0.04),
                arrowprops=dict(arrowstyle='->', color='#AAD4F5',
                                lw=2.0, mutation_scale=18), zorder=5)

# --- Loop-back arrow: TERMINATION -> FITNESS EVALUATION (NO path) ---
term_cx, term_cy = steps[5][0], steps[5][1]
fit_cx,  fit_cy  = steps[1][0], steps[1][1]

loop_x = 8.6   # right side x for the loop line

# Right edge of TERMINATION box
ax.annotate("", xy=(loop_x, term_cy),
            xytext=(term_cx + BOX_W / 2, term_cy),
            arrowprops=dict(arrowstyle='-', color='#F39C12', lw=2.0), zorder=5)

# Vertical line up
ax.plot([loop_x, loop_x], [term_cy, fit_cy], color='#F39C12', lw=2.0, zorder=5)

# Right edge -> FITNESS EVALUATION box (horizontal, with arrowhead)
ax.annotate("", xy=(fit_cx + BOX_W / 2, fit_cy),
            xytext=(loop_x, fit_cy),
            arrowprops=dict(arrowstyle='->', color='#F39C12',
                            lw=2.0, mutation_scale=18), zorder=5)

# "NO (loop)" label
mid_loop_y = (term_cy + fit_cy) / 2
ax.text(loop_x + 0.18, mid_loop_y, "NO\n(loop back)", ha='left', va='center',
        fontsize=9, color='#F39C12', fontweight='bold', zorder=6)

# --- YES output at bottom ---
ax.annotate("", xy=(5, 1.7), xytext=(5, steps[5][1] - BOX_H / 2 - 0.04),
            arrowprops=dict(arrowstyle='->', color='#58D68D',
                            lw=2.0, mutation_scale=18), zorder=5)

yes_box = FancyBboxPatch((5 - 2.8, 1.1), 5.6, 0.55,
                          boxstyle="round,pad=0.08",
                          linewidth=1.2, edgecolor='#58D68D',
                          facecolor='#1D5C38', zorder=3)
ax.add_patch(yes_box)
ax.text(5, 1.38, "YES  →  Output Best Solution", ha='center', va='center',
        fontsize=10, color='#58D68D', fontweight='bold', zorder=4)

ax.set_title("Evolutionary Computation — General Workflow",
             fontsize=14, fontweight='bold', color='white', pad=12)

plt.tight_layout()
plt.savefig("ec_workflow.png", dpi=180, bbox_inches='tight',
            facecolor=fig.get_facecolor())
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns
import numpy as np

limitations = [
    "Curse of\nDimensionality",
    "High\nComputational Cost",
    "Scalability\nIssues",
    "Premature\nConvergence",
    "Parameter\nSensitivity",
    "Lack of Theoretical\nGuarantees",
    "Overfitting in\nOptimization"
]
severity  = [8.5, 8.0, 7.8, 7.5, 7.0, 6.5, 6.0]
solutions = [
    "Dimensionality Reduction / Decomposition",
    "Surrogate Models / Parallel Computing",
    "Distributed EC / Island Model",
    "Diversity Maintenance / Niching",
    "Adaptive Operators / Auto-tuning",
    "Ensemble Approaches / Hybrid Algorithms",
    "Regularization Techniques"
]

palette = sns.color_palette("rocket_r", len(limitations))
y_pos   = np.arange(len(limitations))

fig, ax = plt.subplots(figsize=(13, 6.5))
fig.patch.set_facecolor('#0D1B2A')
ax.set_facecolor('#102030')

bars = ax.barh(y_pos, severity, color=palette,
               edgecolor='#AAD4F5', linewidth=0.8, height=0.58, zorder=3)

# Severity score label at end of each bar
for bar, val, sol in zip(bars, severity, solutions):
    ax.text(val + 0.12, bar.get_y() + bar.get_height() / 2,
            f"{val}/10", va='center', ha='left',
            fontsize=9.5, color='white', fontweight='bold')
    # Solution hint inside bar
    ax.text(0.18, bar.get_y() + bar.get_height() / 2,
            f"↳ {sol}", va='center', ha='left',
            fontsize=7.8, color='#D6EAF8', style='italic', zorder=4)

ax.set_yticks(y_pos)
ax.set_yticklabels(limitations, fontsize=10, color='#D6EAF8')
ax.set_xlim(0, 11.5)
ax.set_xlabel("Impact Severity Score (1–10)", fontsize=11, color='white', labelpad=8)
ax.set_title("Key Limitations of Evolutionary Computation — Severity & Proposed Solutions",
             fontsize=13, fontweight='bold', color='white', pad=12)
ax.invert_yaxis()
ax.tick_params(axis='x', colors='#AABBC5')
ax.tick_params(axis='y', length=0)

# Threshold lines
ax.axvline(x=7.0, color='#F39C12', lw=1.2, linestyle='--', alpha=0.6, zorder=2)
ax.text(7.05, -0.55, "High impact\nthreshold", fontsize=7.5,
        color='#F39C12', va='top', ha='left')

for spine in ['top', 'right']:
    ax.spines[spine].set_visible(False)
for spine in ['bottom', 'left']:
    ax.spines[spine].set_color('#AAD4F5')

plt.tight_layout()
plt.savefig("ec_limitations.png", dpi=180, bbox_inches='tight',
            facecolor=fig.get_facecolor())
plt.show()
```
