```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

# Generate some random 3D data
np.random.seed(42)  # for reproducibility
x = np.random.rand(50)
y = np.random.rand(50)
z = np.random.rand(50)

# Create a new figure for 3D plotting
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot the data as a scatter plot
ax.scatter(x, y, z, c='blue', marker='o')

# Set labels
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')

# Show the plot
plt.title("3D Scatter Plot Example")
plt.show()

```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import random

# Simulated drone positions
drone_positions = {
    'drone1': [0, 0, 5],
    'drone2': [2, 2, 6],
    'drone3': [-1, 3, 4],
}

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Update loop
for i in range(100):  # simulate 100 time steps
    ax.clear()
    ax.set_xlim(-10, 10)
    ax.set_ylim(-10, 10)
    ax.set_zlim(0, 10)
    ax.set_title("Swarm Drone 3D Visualization")

    for drone_id, pos in drone_positions.items():
        # Simulate random movement
        pos[0] += random.uniform(-0.2, 0.2)
        pos[1] += random.uniform(-0.2, 0.2)
        pos[2] += random.uniform(-0.1, 0.1)

        ax.scatter(pos[0], pos[1], pos[2], label=drone_id)

    plt.legend()
    plt.pause(0.1)
    
```
